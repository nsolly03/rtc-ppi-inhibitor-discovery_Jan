## Section 1: Setup and Imports

```python
# Suppress warnings
import warnings
warnings.filterwarnings('ignore')

# Import libraries
import py3Dmol
from Bio import PDB
import numpy as np
import pandas as pd
import json
from pathlib import Path
from datetime import datetime

# Configure paths
PDB_DIR = Path('data/structures/pdb')
RESULTS_DIR = Path('data/analysis_results')
PDB_FILE = PDB_DIR / '6XEZ.pdb'

# Verify setup
RESULTS_DIR.mkdir(exist_ok=True, parents=True)
assert PDB_FILE.exists(), f"PDB file not found: {PDB_FILE}"

print("✓ Setup complete")
print(f"  PDB file: {PDB_FILE.resolve()}")
print(f"  Results: {RESULTS_DIR.resolve()}")
print()
print("⚠️  COMPLEX STRUCTURE: RTC-Helicase Supercomplex")
print("   5 proteins + RNA = Multiple interfaces to analyze")
print()
print("🎯 PRIMARY FOCUS: NSP12-NSP13 interface (unique to this structure)")
```

## Section 2: Load and Parse Structure

```python
# Parse structure
parser = PDB.PDBParser(QUIET=True)
structure = parser.get_structure('6XEZ', str(PDB_FILE))
model = structure[0]

print("="*70)
print("STRUCTURE INFORMATION - 6XEZ")
print("="*70)
print()

# Collect all chain info
all_chains = []
for chain in model:
    residues = [r for r in chain.get_residues() if r.id[0] == ' ']
    num_res = len(residues)
    
    if residues:
        first_res = residues[0].id[1]
        last_res = residues[-1].id[1]
    else:
        first_res = last_res = 'N/A'
    
    all_chains.append({
        'id': chain.id,
        'num_res': num_res,
        'first': first_res,
        'last': last_res
    })
    
    print(f"Chain {chain.id}: {num_res} residues (PDB {first_res}-{last_res})")

print()
print("Analyzing chain sizes...")
print()

# Identify chains (based on residue counts)
chain_assignments = {}
nsp12_chains = []
nsp7_chains = []
nsp8_chains = []
nsp13_chains = []
rna_chains = []

for chain_data in all_chains:
    num_res = chain_data['num_res']
    chain_id = chain_data['id']
    
    # NSP12 (RdRp): ~900-950 residues
    if 850 < num_res < 1000:
        protein = 'NSP12 (RdRp)'
        nsp12_chains.append(chain_id)
    # NSP13 (helicase): ~570-600 residues
    elif 550 < num_res < 620:
        protein = 'NSP13 (helicase)'
        nsp13_chains.append(chain_id)
    # NSP7: ~60-90 residues
    elif 60 < num_res < 90:
        protein = 'NSP7 (cofactor)'
        nsp7_chains.append(chain_id)
    # NSP8: ~180-200 residues
    elif 180 < num_res < 210:
        protein = 'NSP8 (cofactor)'
        nsp8_chains.append(chain_id)
    # RNA or small ligand
    elif num_res < 50:
        protein = 'RNA/Ligand'
        rna_chains.append(chain_id)
    else:
        protein = 'Unknown'
    
    print(f"Chain {chain_id} ({num_res} res) → {protein}")

print()
print("="*70)
print("FINAL ASSIGNMENTS:")
print("="*70)

if nsp12_chains:
    chain_assignments['nsp12'] = nsp12_chains[0]
    print(f"  NSP12 (RdRp) = Chain {nsp12_chains[0]}")

if nsp7_chains:
    chain_assignments['nsp7'] = nsp7_chains[0]
    print(f"  NSP7 (cofactor) = Chain {nsp7_chains[0]}")

if nsp8_chains:
    chain_assignments['nsp8'] = nsp8_chains[0]
    print(f"  NSP8 (cofactor) = Chain {nsp8_chains[0]}")

if len(nsp13_chains) >= 1:
    chain_assignments['nsp13_1'] = nsp13_chains[0]
    print(f"  NSP13 Copy 1 = Chain {nsp13_chains[0]}")
    
if len(nsp13_chains) >= 2:
    chain_assignments['nsp13_2'] = nsp13_chains[1]
    print(f"  NSP13 Copy 2 = Chain {nsp13_chains[1]}")

if rna_chains:
    print(f"  RNA/Ligands = Chains {', '.join(rna_chains)}")

print()
print("Target Interfaces:")
print("  PRIMARY: NSP12-NSP13 (unique to this structure!)")
if len(nsp13_chains) >= 2:
    print("  SECONDARY: NSP13-NSP13 (compare with 6ZSL)")
print("  TERTIARY: NSP12-NSP7, NSP12-NSP8 (compare with 7DFG)")
print("="*70)
```

## Section 3: Visualize Structure

```python
# Read PDB file
with open(PDB_FILE, 'r') as f:
    pdb_data = f.read()

# Create viewer
view = py3Dmol.view(width=900, height=700)
view.addModel(pdb_data, 'pdb')

# Color by protein type
if 'nsp12' in chain_assignments:
    view.setStyle({'chain': chain_assignments['nsp12']}, 
                  {'cartoon': {'color': 'blue', 'opacity': 0.8}})

if 'nsp7' in chain_assignments:
    view.setStyle({'chain': chain_assignments['nsp7']}, 
                  {'cartoon': {'color': 'green', 'opacity': 0.8}})

if 'nsp8' in chain_assignments:
    view.setStyle({'chain': chain_assignments['nsp8']}, 
                  {'cartoon': {'color': 'yellow', 'opacity': 0.8}})

if 'nsp13_1' in chain_assignments:
    view.setStyle({'chain': chain_assignments['nsp13_1']}, 
                  {'cartoon': {'color': 'red', 'opacity': 0.8}})

if 'nsp13_2' in chain_assignments:
    view.setStyle({'chain': chain_assignments['nsp13_2']}, 
                  {'cartoon': {'color': 'orange', 'opacity': 0.8}})

# RNA/ligands transparent
for rna_chain in rna_chains:
    view.setStyle({'chain': rna_chain}, 
                  {'cartoon': {'color': 'gray', 'opacity': 0.3}})

# Add labels
if 'nsp12' in chain_assignments:
    nsp12 = model[chain_assignments['nsp12']]
    residues = [r for r in nsp12.get_residues() if r.id[0] == ' ']
    if len(residues) > 10:
        mid_res = residues[len(residues)//2].id[1]
        view.addLabel(f"NSP12 (RdRp)", 
                      {'fontColor': 'white', 'fontSize': 14, 
                       'backgroundColor': 'blue', 'backgroundOpacity': 0.8},
                      {'chain': chain_assignments['nsp12'], 'resi': mid_res})

if 'nsp13_1' in chain_assignments:
    nsp13 = model[chain_assignments['nsp13_1']]
    residues = [r for r in nsp13.get_residues() if r.id[0] == ' ']
    if len(residues) > 10:
        mid_res = residues[len(residues)//2].id[1]
        view.addLabel(f"NSP13 (Helicase)", 
                      {'fontColor': 'white', 'fontSize': 14, 
                       'backgroundColor': 'red', 'backgroundOpacity': 0.8},
                      {'chain': chain_assignments['nsp13_1'], 'resi': mid_res})

view.zoomTo()
view.setBackgroundColor('white')

print("3D Structure - RTC-Helicase Supercomplex:")
print("  Blue = NSP12 (RdRp)")
print("  Green = NSP7")
print("  Yellow = NSP8")
print("  Red = NSP13 Copy 1")
print("  Orange = NSP13 Copy 2")
print("  Gray = RNA/Ligands")
print()
print("🎯 Focus: Blue-Red interface (NSP12-NSP13)")

view.show()
```

## Section 4: DISCOVER NSP12-NSP13 Hot Spots (PRIMARY FOCUS)

```python
print("="*70)
print("DISCOVERING HOT SPOTS: NSP12-NSP13 INTERFACE")
print("="*70)
print()
print("🎯 PRIMARY ANALYSIS - This interface is UNIQUE to 6XEZ!")
print()

if 'nsp12' not in chain_assignments or 'nsp13_1' not in chain_assignments:
    print("⚠️  Missing required chains!")
    df_hot_nsp12_nsp13 = pd.DataFrame()
else:
    nsp12 = model[chain_assignments['nsp12']]
    nsp13 = model[chain_assignments['nsp13_1']]
    
    # Get all atoms
    nsp12_atoms = [atom for res in nsp12 if res.id[0] == ' ' for atom in res.get_atoms()]
    nsp13_atoms = [atom for res in nsp13 if res.id[0] == ' ' for atom in res.get_atoms()]
    
    print(f"NSP12 (Chain {chain_assignments['nsp12']}): {len(nsp12_atoms)} atoms")
    print(f"NSP13 (Chain {chain_assignments['nsp13_1']}): {len(nsp13_atoms)} atoms")
    print()
    print("Searching for ALL charged interactions (< 5 Å)...")
    print("This may take a minute (large proteins!)...")
    print()
    
    # Find ALL interface contacts < 5 Å
    interface_contacts = []
    for atom1 in nsp12_atoms:
        for atom2 in nsp13_atoms:
            dist = np.linalg.norm(atom1.get_coord() - atom2.get_coord())
            if dist < 5.0:
                res1 = atom1.get_parent()
                res2 = atom2.get_parent()
                interface_contacts.append((res1, res2, dist, atom1.get_name(), atom2.get_name()))
    
    print(f"✓ Found {len(interface_contacts)} atomic contacts < 5 Å")
    print()
    
    # Get unique residue pairs
    unique_pairs = {}
    for res1, res2, dist, atom1, atom2 in interface_contacts:
        key = (res1.id[1], res2.id[1])
        if key not in unique_pairs or dist < unique_pairs[key][2]:
            unique_pairs[key] = (res1, res2, dist, atom1, atom2)
    
    print(f"✓ Unique residue pairs: {len(unique_pairs)}")
    print()
    
    # DISCOVER ALL charged interactions
    print("="*70)
    print("CHARGED INTERACTION DISCOVERY")
    print("="*70)
    print()
    
    charged_positive = ['LYS', 'ARG']
    charged_negative = ['ASP', 'GLU']
    
    hot_spots_nsp12_nsp13 = []
    for (pdb1, pdb2), (res1, res2, dist, atom1, atom2) in unique_pairs.items():
        resname1 = res1.get_resname()
        resname2 = res2.get_resname()
        
        # Check for charged interactions
        if (resname1 in charged_positive and resname2 in charged_negative) or \
           (resname1 in charged_negative and resname2 in charged_positive):
            
            interaction_type = 'Salt bridge' if dist < 4.0 else 'Ionic interaction'
            
            hot_spots_nsp12_nsp13.append({
                'NSP12_Res': resname1,
                'NSP12_PDB': pdb1,
                'NSP13_Res': resname2,
                'NSP13_PDB': pdb2,
                'Distance': dist,
                'Atom1': atom1,
                'Atom2': atom2,
                'Type': interaction_type
            })
    
    if hot_spots_nsp12_nsp13:
        df_hot_nsp12_nsp13 = pd.DataFrame(hot_spots_nsp12_nsp13).sort_values('Distance')
        
        print(f"✓ DISCOVERED {len(df_hot_nsp12_nsp13)} charged interactions!")
        print()
        print("All Charged Interactions (sorted by distance):")
        print()
        print(df_hot_nsp12_nsp13.to_string(index=False))
        print()
        
        # Highlight strongest
        strongest = df_hot_nsp12_nsp13.iloc[0]
        print("="*70)
        print("🔥 STRONGEST HOT SPOT (NSP12-NSP13):")
        print(f"  {strongest['NSP12_Res']}{strongest['NSP12_PDB']} (NSP12) ←→ {strongest['NSP13_Res']}{strongest['NSP13_PDB']} (NSP13)")
        print(f"  Distance: {strongest['Distance']:.2f} Å")
        print(f"  Type: {strongest['Type']}")
        print("="*70)
        print()
        print("🎯 This is a UNIQUE interface - not found in 7DFG or 6ZSL!")
        print("   RdRp-Helicase coupling could be critical for replication")
    else:
        print("⚠️  No charged interactions found < 5 Å")
        df_hot_nsp12_nsp13 = pd.DataFrame()

print()
```

## Section 5: Visualize NSP12-NSP13 Hot Spot

```python
# Visualize NSP12-NSP13 hot spot
if not df_hot_nsp12_nsp13.empty and 'nsp12' in chain_assignments and 'nsp13_1' in chain_assignments:
    with open(PDB_FILE, 'r') as f:
        pdb_data = f.read()
    
    view = py3Dmol.view(width=800, height=600)
    view.addModel(pdb_data, 'pdb')
    
    # Show proteins transparent
    view.setStyle({'chain': chain_assignments['nsp12']}, 
                  {'cartoon': {'color': 'blue', 'opacity': 0.3}})
    view.setStyle({'chain': chain_assignments['nsp13_1']}, 
                  {'cartoon': {'color': 'red', 'opacity': 0.3}})
    
    # Get hot spot residues
    top = df_hot_nsp12_nsp13.iloc[0]
    
    # Convert to Python int
    res_nsp12_pdb = int(top['NSP12_PDB'])
    res_nsp13_pdb = int(top['NSP13_PDB'])
    
    # Highlight as SPHERES
    view.addStyle({'chain': chain_assignments['nsp12'], 'resi': res_nsp12_pdb},
                  {'sphere': {'color': 'blue', 'radius': 1.0}})
    
    view.addStyle({'chain': chain_assignments['nsp13_1'], 'resi': res_nsp13_pdb},
                  {'sphere': {'color': 'red', 'radius': 1.0}})
    
    # Add labels with distance
    view.addLabel(f"{top['NSP12_Res']}{res_nsp12_pdb} (NSP12)",
                  {'fontColor': 'white', 'fontSize': 16, 
                   'backgroundColor': 'blue', 'backgroundOpacity': 0.8},
                  {'chain': chain_assignments['nsp12'], 'resi': res_nsp12_pdb})
    
    view.addLabel(f"{top['NSP13_Res']}{res_nsp13_pdb} (NSP13) {top['Distance']:.1f} Å",
                  {'fontColor': 'white', 'fontSize': 16, 
                   'backgroundColor': 'red', 'backgroundOpacity': 0.8},
                  {'chain': chain_assignments['nsp13_1'], 'resi': res_nsp13_pdb})
    
    # Center on hot spot
    view.zoomTo({'chain': chain_assignments['nsp12'], 'resi': res_nsp12_pdb})
    view.setBackgroundColor('white')
    
    print(f"Hot Spot: {top['NSP12_Res']}{res_nsp12_pdb} (NSP12) ←→ {top['NSP13_Res']}{res_nsp13_pdb} (NSP13)")
    print(f"Distance: {top['Distance']:.2f} Å")
    print()
    print("Blue sphere = NSP12 (RdRp)")
    print("Red sphere = NSP13 (Helicase)")
    print()
    print("🎯 RdRp-Helicase coupling interface!")
    
    view.show()
else:
    print("No NSP12-NSP13 hot spots to visualize")
```

## Section 6: Quick Check - NSP13-NSP13 Interface (Compare with 6ZSL)

```python
print("="*70)
print("QUICK CHECK: NSP13-NSP13 INTERFACE (Compare with 6ZSL)")
print("="*70)
print()

if 'nsp13_1' not in chain_assignments or 'nsp13_2' not in chain_assignments:
    print("⚠️  Only one NSP13 chain found - cannot analyze homodimer")
    df_hot_nsp13_nsp13 = pd.DataFrame()
else:
    print(f"Analyzing NSP13-NSP13 interface (Chains {chain_assignments['nsp13_1']}-{chain_assignments['nsp13_2']})")
    print("This is a QUICK check - we already analyzed this in 6ZSL")
    print()
    
    nsp13_a = model[chain_assignments['nsp13_1']]
    nsp13_b = model[chain_assignments['nsp13_2']]
    
    # Quick scan for charged interactions < 5 Å
    nsp13_a_atoms = [atom for res in nsp13_a if res.id[0] == ' ' for atom in res.get_atoms()]
    nsp13_b_atoms = [atom for res in nsp13_b if res.id[0] == ' ' for atom in res.get_atoms()]
    
    interface_contacts = []
    for atom1 in nsp13_a_atoms:
        for atom2 in nsp13_b_atoms:
            dist = np.linalg.norm(atom1.get_coord() - atom2.get_coord())
            if dist < 5.0:
                res1 = atom1.get_parent()
                res2 = atom2.get_parent()
                interface_contacts.append((res1, res2, dist, atom1.get_name(), atom2.get_name()))
    
    # Get unique pairs
    unique_pairs = {}
    for res1, res2, dist, atom1, atom2 in interface_contacts:
        key = (res1.id[1], res2.id[1])
        if key not in unique_pairs or dist < unique_pairs[key][2]:
            unique_pairs[key] = (res1, res2, dist, atom1, atom2)
    
    # Find charged interactions
    charged_positive = ['LYS', 'ARG']
    charged_negative = ['ASP', 'GLU']
    
    hot_spots_nsp13_nsp13 = []
    for (pdb1, pdb2), (res1, res2, dist, atom1, atom2) in unique_pairs.items():
        resname1 = res1.get_resname()
        resname2 = res2.get_resname()
        
        if (resname1 in charged_positive and resname2 in charged_negative) or \
           (resname1 in charged_negative and resname2 in charged_positive):
            
            hot_spots_nsp13_nsp13.append({
                'NSP13_A_Res': resname1,
                'NSP13_A_PDB': pdb1,
                'NSP13_B_Res': resname2,
                'NSP13_B_PDB': pdb2,
                'Distance': dist
            })
    
    if hot_spots_nsp13_nsp13:
        df_hot_nsp13_nsp13 = pd.DataFrame(hot_spots_nsp13_nsp13).sort_values('Distance')
        
        print(f"✓ Found {len(df_hot_nsp13_nsp13)} NSP13-NSP13 hot spots")
        print()
        print(df_hot_nsp13_nsp13.head(3).to_string(index=False))
        print()
        
        strongest = df_hot_nsp13_nsp13.iloc[0]
        print(f"Strongest: {strongest['NSP13_A_Res']}{strongest['NSP13_A_PDB']}-{strongest['NSP13_B_Res']}{strongest['NSP13_B_PDB']} ({strongest['Distance']:.2f} Å)")
        print()
        print("📊 COMPARISON WITH 6ZSL:")
        print("  6ZSL strongest: ARG390-GLU365 (3.05 Å)")
        print(f"  6XEZ strongest: {strongest['NSP13_A_Res']}{strongest['NSP13_A_PDB']}-{strongest['NSP13_B_Res']}{strongest['NSP13_B_PDB']} ({strongest['Distance']:.2f} Å)")
        print()
        if abs(strongest['Distance'] - 3.05) < 0.5:
            print("  ✓ Similar to 6ZSL - confirms homodimer interface")
        else:
            print("  ⚠️ Different from 6ZSL - may be different conformation")
    else:
        print("⚠️  No NSP13-NSP13 charged interactions < 5 Å found")
        df_hot_nsp13_nsp13 = pd.DataFrame()

print()
print("="*70)
```

## Section 7: Summary and Export

```python
print("="*70)
print("6XEZ ANALYSIS SUMMARY")
print("RTC-Helicase Supercomplex")
print("="*70)
print()

# Summary data
summary_data = []
summary_data.append(['Property', 'Value'])
summary_data.append(['PDB ID', '6XEZ'])
summary_data.append(['Analysis Date', datetime.now().strftime('%Y-%m-%d')])
summary_data.append(['Structure Type', 'Supercomplex (5 proteins + RNA)'])
summary_data.append(['', ''])

summary_data.append(['COMPOSITION', ''])
if 'nsp12' in chain_assignments:
    summary_data.append(['NSP12 (RdRp)', f"Chain {chain_assignments['nsp12']}"])
if 'nsp7' in chain_assignments:
    summary_data.append(['NSP7', f"Chain {chain_assignments['nsp7']}"])
if 'nsp8' in chain_assignments:
    summary_data.append(['NSP8', f"Chain {chain_assignments['nsp8']}"])
if 'nsp13_1' in chain_assignments:
    summary_data.append(['NSP13 Copy 1', f"Chain {chain_assignments['nsp13_1']}"])
if 'nsp13_2' in chain_assignments:
    summary_data.append(['NSP13 Copy 2', f"Chain {chain_assignments['nsp13_2']}"])
summary_data.append(['', ''])

summary_data.append(['NSP12-NSP13 INTERFACE', ''])
if not df_hot_nsp12_nsp13.empty:
    top = df_hot_nsp12_nsp13.iloc[0]
    summary_data.append(['Hot Spots Found', str(len(df_hot_nsp12_nsp13))])
    summary_data.append(['Strongest', f"{top['NSP12_Res']}{int(top['NSP12_PDB'])}-{top['NSP13_Res']}{int(top['NSP13_PDB'])}"])
    summary_data.append(['Distance', f"{top['Distance']:.2f} Å"])
    summary_data.append(['Status', 'UNIQUE to 6XEZ ⭐'])
else:
    summary_data.append(['Status', 'None found'])
summary_data.append(['', ''])

summary_data.append(['NSP13-NSP13 INTERFACE', ''])
if not df_hot_nsp13_nsp13.empty:
    summary_data.append(['Hot Spots Found', str(len(df_hot_nsp13_nsp13))])
    summary_data.append(['Status', 'Compare with 6ZSL'])
else:
    summary_data.append(['Status', 'None found'])

df_summary = pd.DataFrame(summary_data[1:], columns=summary_data[0])

print(df_summary.to_string(index=False))
print()
print("="*70)
print()

# Export results
print("Exporting results...")
print()

if not df_hot_nsp12_nsp13.empty:
    csv_file = RESULTS_DIR / '6XEZ_NSP12-NSP13_hotspots.csv'
    df_hot_nsp12_nsp13.to_csv(csv_file, index=False)
    print(f"✓ NSP12-NSP13 hot spots: {csv_file}")

if not df_hot_nsp13_nsp13.empty:
    csv_file = RESULTS_DIR / '6XEZ_NSP13-NSP13_hotspots.csv'
    df_hot_nsp13_nsp13.to_csv(csv_file, index=False)
    print(f"✓ NSP13-NSP13 hot spots: {csv_file}")

# Export JSON
analysis_data = {
    'pdb_id': '6XEZ',
    'analysis_date': datetime.now().strftime('%Y-%m-%d'),
    'structure_type': 'supercomplex',
    'composition': chain_assignments,
    'interfaces': {}
}

if not df_hot_nsp12_nsp13.empty:
    top = df_hot_nsp12_nsp13.iloc[0]
    analysis_data['interfaces']['nsp12_nsp13'] = {
        'hot_spots_count': len(df_hot_nsp12_nsp13),
        'strongest': {
            'nsp12_residue': f"{top['NSP12_Res']}{int(top['NSP12_PDB'])}",
            'nsp13_residue': f"{top['NSP13_Res']}{int(top['NSP13_PDB'])}",
            'distance': float(top['Distance']),
            'type': top['Type']
        },
        'note': 'UNIQUE to 6XEZ - RdRp-Helicase coupling'
    }

if not df_hot_nsp13_nsp13.empty:
    analysis_data['interfaces']['nsp13_nsp13'] = {
        'hot_spots_count': len(df_hot_nsp13_nsp13),
        'note': 'Compare with 6ZSL homodimer'
    }

json_file = RESULTS_DIR / '6XEZ_analysis.json'
with open(json_file, 'w') as f:
    json.dump(analysis_data, f, indent=2)
print(f"✓ Analysis data: {json_file}")

# Export summary
summary_file = RESULTS_DIR / '6XEZ_summary.csv'
df_summary.to_csv(summary_file, index=False)
print(f"✓ Summary table: {summary_file}")

print()
print("✓ All results exported!")
print()
print(f"Results directory: {RESULTS_DIR.resolve()}")
```
