## Key Differences: Homodimer vs Heterodimer

**Heterodimer (what we've done):**
- Two different proteins (e.g., NSP10 + NSP16)
- Asymmetric interface
- One-directional interactions

**Homodimer (this analysis):**
- Same protein, two copies (NSP13 + NSP13)
- **Symmetric interface**
- **Potential duplicate interactions** (K100-D200 = D100-K200)
- Need to avoid counting same interaction twice

---

## Research Questions

1. **What holds the NSP13 dimer together?**
2. **Are there symmetric hot spots?**
3. **Can we disrupt dimerization?**
4. **Is the interface druggable?**

## Section 1: Setup and Imports

```python
# Suppress warnings
import warnings
warnings.filterwarnings('ignore')

# Import libraries
import py3Dmol
from Bio import PDB
import numpy as np
import pandas as pd
import json
import os
from pathlib import Path
from datetime import datetime

# Configure paths
PDB_DIR = Path('data/structures/pdb')
RESULTS_DIR = Path('data/analysis_results')
PDB_FILE = PDB_DIR / '6ZSL.pdb'

# Verify setup
RESULTS_DIR.mkdir(exist_ok=True, parents=True)
assert PDB_FILE.exists(), f"PDB file not found: {PDB_FILE}"

print("✓ Setup complete")
print(f"  PDB file: {PDB_FILE.resolve()}")
print(f"  Results: {RESULTS_DIR.resolve()}")
print(f"  File exists: {PDB_FILE.exists()}")
print()
print("⚠️  HOMODIMER MODE: NSP13-NSP13 (symmetric interface)")
print("   Will check for duplicate interactions")
```

## Section 2: Load and Parse Structure

```python
# Parse structure
parser = PDB.PDBParser(QUIET=True)
structure = parser.get_structure('6ZSL', str(PDB_FILE))
model = structure[0]

print("="*70)
print("STRUCTURE INFORMATION - 6ZSL")
print("="*70)
print()

# Collect all chain info
all_chains = []
for chain in model:
    residues = [r for r in chain.get_residues() if r.id[0] == ' ']
    num_res = len(residues)
    
    if residues:
        first_res = residues[0].id[1]
        last_res = residues[-1].id[1]
    else:
        first_res = last_res = 'N/A'
    
    all_chains.append({
        'id': chain.id,
        'num_res': num_res,
        'first': first_res,
        'last': last_res
    })
    
    print(f"Chain {chain.id}: {num_res} residues (PDB {first_res}-{last_res})")

print()
print("Analyzing chain sizes...")
print()

# Identify NSP13 chains (helicase ~585-590 residues)
chain_assignments = {}
nsp13_chains = []

for chain_data in all_chains:
    num_res = chain_data['num_res']
    chain_id = chain_data['id']
    
    # NSP13 helicase detection (570-600 residues)
    if 570 < num_res < 600:
        protein = 'NSP13 (helicase)'
        nsp13_chains.append(chain_id)
    else:
        protein = 'Unknown/Ligand'
    
    print(f"Chain {chain_id} ({num_res} res) → {protein}")

print()
print("="*70)
print("FINAL ASSIGNMENTS:")
print("="*70)

if len(nsp13_chains) >= 2:
    chain_assignments['nsp13_A'] = nsp13_chains[0]
    chain_assignments['nsp13_B'] = nsp13_chains[1]
    print(f"  NSP13 Copy A = Chain {nsp13_chains[0]}")
    print(f"  NSP13 Copy B = Chain {nsp13_chains[1]}")
    print()
    print("Target Interface:")
    print(f"  NSP13-NSP13 homodimer (Chain {nsp13_chains[0]} - Chain {nsp13_chains[1]})")
    print()
    print("⚠️  HOMODIMER: May see symmetric interactions")
else:
    print("  ❌ ERROR: Expected 2 NSP13 chains, found:", len(nsp13_chains))

print("="*70)
```

## Section 3: Visualize Structure (py3Dmol)

```python
# Read PDB file
with open(PDB_FILE, 'r') as f:
    pdb_data = f.read()

# Create viewer
view = py3Dmol.view(width=800, height=600)
view.addModel(pdb_data, 'pdb')

# Color by chain (different colors for each NSP13 copy)
if len(nsp13_chains) >= 2:
    view.setStyle({'chain': nsp13_chains[0]}, {'cartoon': {'color': 'cyan', 'opacity': 0.8}})
    view.setStyle({'chain': nsp13_chains[1]}, {'cartoon': {'color': 'magenta', 'opacity': 0.8}})
    
    # Add labels
    nsp13_A = model[nsp13_chains[0]]
    residues_A = [r for r in nsp13_A.get_residues() if r.id[0] == ' ']
    if len(residues_A) > 10:
        mid_res_A = residues_A[len(residues_A)//2].id[1]
        view.addLabel(f"NSP13-A (Chain {nsp13_chains[0]})", 
                      {'fontColor': 'white', 'fontSize': 14, 'backgroundColor': 'cyan', 'backgroundOpacity': 0.8},
                      {'chain': nsp13_chains[0], 'resi': mid_res_A})
    
    nsp13_B = model[nsp13_chains[1]]
    residues_B = [r for r in nsp13_B.get_residues() if r.id[0] == ' ']
    if len(residues_B) > 10:
        mid_res_B = residues_B[len(residues_B)//2].id[1]
        view.addLabel(f"NSP13-B (Chain {nsp13_chains[1]})", 
                      {'fontColor': 'white', 'fontSize': 14, 'backgroundColor': 'magenta', 'backgroundOpacity': 0.8},
                      {'chain': nsp13_chains[1], 'resi': mid_res_B})

view.zoomTo()
view.setBackgroundColor('white')

print("3D Structure - NSP13 Helicase Homodimer:")
if len(nsp13_chains) >= 2:
    print(f"  Cyan = NSP13 Copy A (Chain {nsp13_chains[0]})")
    print(f"  Magenta = NSP13 Copy B (Chain {nsp13_chains[1]})")
print()
print("Note: Same protein, two copies (homodimer)")

view.show()
```

## Section 4: DISCOVER NSP13-NSP13 Hot Spots (Homodimer)

```python
print("="*70)
print("DISCOVERING HOT SPOTS: NSP13-NSP13 HOMODIMER INTERFACE")
print("="*70)
print()

if len(nsp13_chains) < 2:
    print("⚠️  Missing required chains!")
    df_hot_nsp13 = pd.DataFrame()
else:
    nsp13_A = model[nsp13_chains[0]]
    nsp13_B = model[nsp13_chains[1]]
    
    # Get all atoms
    nsp13_A_atoms = [atom for res in nsp13_A if res.id[0] == ' ' for atom in res.get_atoms()]
    nsp13_B_atoms = [atom for res in nsp13_B if res.id[0] == ' ' for atom in res.get_atoms()]
    
    print(f"NSP13 Copy A (Chain {nsp13_chains[0]}): {len(nsp13_A_atoms)} atoms")
    print(f"NSP13 Copy B (Chain {nsp13_chains[1]}): {len(nsp13_B_atoms)} atoms")
    print()
    print("Searching for ALL charged interactions (< 5 Å)...")
    print("This may take a minute...")
    print()
    
    # Find ALL interface contacts < 5 Å
    interface_contacts = []
    for atom1 in nsp13_A_atoms:
        for atom2 in nsp13_B_atoms:
            dist = np.linalg.norm(atom1.get_coord() - atom2.get_coord())
            if dist < 5.0:
                res1 = atom1.get_parent()
                res2 = atom2.get_parent()
                interface_contacts.append((res1, res2, dist, atom1.get_name(), atom2.get_name()))
    
    print(f"✓ Found {len(interface_contacts)} atomic contacts < 5 Å")
    print()
    
    # Get unique residue pairs
    unique_pairs = {}
    for res1, res2, dist, atom1, atom2 in interface_contacts:
        key = (res1.id[1], res2.id[1])
        if key not in unique_pairs or dist < unique_pairs[key][2]:
            unique_pairs[key] = (res1, res2, dist, atom1, atom2)
    
    print(f"✓ Unique residue pairs: {len(unique_pairs)}")
    print()
    
    # DISCOVER ALL charged interactions
    print("="*70)
    print("CHARGED INTERACTION DISCOVERY")
    print("="*70)
    print()
    
    charged_positive = ['LYS', 'ARG']
    charged_negative = ['ASP', 'GLU']
    
    hot_spots_nsp13 = []
    for (pdb1, pdb2), (res1, res2, dist, atom1, atom2) in unique_pairs.items():
        resname1 = res1.get_resname()
        resname2 = res2.get_resname()
        
        # Check for charged interactions
        if (resname1 in charged_positive and resname2 in charged_negative) or \
           (resname1 in charged_negative and resname2 in charged_positive):
            
            interaction_type = 'Salt bridge' if dist < 4.0 else 'Ionic interaction'
            
            hot_spots_nsp13.append({
                'NSP13_A_Res': resname1,
                'NSP13_A_PDB': pdb1,
                'NSP13_B_Res': resname2,
                'NSP13_B_PDB': pdb2,
                'Distance': dist,
                'Atom1': atom1,
                'Atom2': atom2,
                'Type': interaction_type
            })
    
    if hot_spots_nsp13:
        df_hot_nsp13 = pd.DataFrame(hot_spots_nsp13).sort_values('Distance')
        
        print(f"✓ DISCOVERED {len(df_hot_nsp13)} charged interactions!")
        print()
        print("All Charged Interactions (sorted by distance):")
        print()
        print(df_hot_nsp13.to_string(index=False))
        print()
        
        # Highlight strongest interaction
        strongest = df_hot_nsp13.iloc[0]
        print("="*70)
        print("🔥 STRONGEST HOT SPOT (NSP13-NSP13):")
        print(f"  {strongest['NSP13_A_Res']}{strongest['NSP13_A_PDB']} (Chain {nsp13_chains[0]}) ←→ {strongest['NSP13_B_Res']}{strongest['NSP13_B_PDB']} (Chain {nsp13_chains[1]})")
        print(f"  Distance: {strongest['Distance']:.2f} Å")
        print(f"  Type: {strongest['Type']}")
        print("="*70)
        print()
        
        # Check for symmetric pairs
        print("🔍 CHECKING FOR SYMMETRIC INTERACTIONS:")
        print()
        symmetric_found = False
        for i, row1 in df_hot_nsp13.iterrows():
            # Look for reverse pair (A-B vs B-A with swapped residue numbers)
            for j, row2 in df_hot_nsp13.iterrows():
                if i < j:  # Avoid self-comparison
                    if (row1['NSP13_A_PDB'] == row2['NSP13_B_PDB'] and 
                        row1['NSP13_B_PDB'] == row2['NSP13_A_PDB']):
                        print(f"  Symmetric pair found:")
                        print(f"    {row1['NSP13_A_Res']}{row1['NSP13_A_PDB']}-{row1['NSP13_B_Res']}{row1['NSP13_B_PDB']}")
                        print(f"    {row2['NSP13_A_Res']}{row2['NSP13_A_PDB']}-{row2['NSP13_B_Res']}{row2['NSP13_B_PDB']}")
                        symmetric_found = True
        
        if not symmetric_found:
            print("  No perfect symmetric pairs found")
            print("  (Interface may be pseudo-symmetric or asymmetric)")
        print("="*70)
    else:
        print("⚠️  No charged interactions found < 5 Å")
        df_hot_nsp13 = pd.DataFrame()

print()
```

## Section 5: Visualize NSP13-NSP13 Hot Spot

```python
# Visualize hot spot
if not df_hot_nsp13.empty and len(nsp13_chains) >= 2:
    with open(PDB_FILE, 'r') as f:
        pdb_data = f.read()

    view = py3Dmol.view(width=800, height=600)
    view.addModel(pdb_data, 'pdb')

    view.setStyle({'chain': nsp13_chains[0]}, {'cartoon': {'color': 'cyan', 'opacity': 0.3}})
    view.setStyle({'chain': nsp13_chains[1]}, {'cartoon': {'color': 'magenta', 'opacity': 0.3}})

    top = df_hot_nsp13.iloc[0]

    view.addStyle({'chain': nsp13_chains[0], 'resi': top['NSP13_A_PDB']},
                  {'stick': {'colorscheme': 'redCarbon', 'radius': 0.3}})
    view.addStyle({'chain': nsp13_chains[1], 'resi': top['NSP13_B_PDB']},
                  {'stick': {'colorscheme': 'blueCarbon', 'radius': 0.3}})

    view.zoomTo()
    view.setBackgroundColor('white')
    view.show()
else:
    print('No hot spots to visualize')
```
