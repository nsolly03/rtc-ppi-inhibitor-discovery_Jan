## Section 1: Setup and Imports

```python
# Suppress warnings
import warnings
warnings.filterwarnings('ignore')

# Import libraries
import py3Dmol
from Bio import PDB
import numpy as np
import pandas as pd
import json
from pathlib import Path
from datetime import datetime

# Configure paths
PDB_DIR = Path('data/structures/pdb')
RESULTS_DIR = Path('data/analysis_results')
PDB_FILE = PDB_DIR / '6ZSL.pdb'

# Verify setup
RESULTS_DIR.mkdir(exist_ok=True, parents=True)
assert PDB_FILE.exists(), f"PDB file not found: {PDB_FILE}"

print("✓ Setup complete")
print(f"  PDB file: {PDB_FILE.resolve()}")
print(f"  Results: {RESULTS_DIR.resolve()}")
print()
print("⚠️  HOMODIMER MODE: NSP13-NSP13 (symmetric interface)")
```

## Section 2: Load and Parse Structure

```python
# Parse structure
parser = PDB.PDBParser(QUIET=True)
structure = parser.get_structure('6ZSL', str(PDB_FILE))
model = structure[0]

print("="*70)
print("STRUCTURE INFORMATION - 6ZSL")
print("="*70)
print()

# Collect all chain info
all_chains = []
for chain in model:
    residues = [r for r in chain.get_residues() if r.id[0] == ' ']
    num_res = len(residues)
    
    if residues:
        first_res = residues[0].id[1]
        last_res = residues[-1].id[1]
    else:
        first_res = last_res = 'N/A'
    
    all_chains.append({
        'id': chain.id,
        'num_res': num_res,
        'first': first_res,
        'last': last_res
    })
    
    print(f"Chain {chain.id}: {num_res} residues (PDB {first_res}-{last_res})")

print()
print("Analyzing chain sizes...")
print()

# Identify NSP13 chains
chain_assignments = {}
nsp13_chains = []

for chain_data in all_chains:
    num_res = chain_data['num_res']
    chain_id = chain_data['id']
    
    if 570 < num_res < 600:
        protein = 'NSP13 (helicase)'
        nsp13_chains.append(chain_id)
    else:
        protein = 'Unknown/Ligand'
    
    print(f"Chain {chain_id} ({num_res} res) → {protein}")

print()
print("="*70)
print("FINAL ASSIGNMENTS:")
print("="*70)

if len(nsp13_chains) >= 2:
    chain_assignments['nsp13_A'] = nsp13_chains[0]
    chain_assignments['nsp13_B'] = nsp13_chains[1]
    print(f"  NSP13 Copy A = Chain {nsp13_chains[0]}")
    print(f"  NSP13 Copy B = Chain {nsp13_chains[1]}")
    print()
    print("Target Interface:")
    print(f"  NSP13-NSP13 homodimer (Chain {nsp13_chains[0]} - Chain {nsp13_chains[1]})")
else:
    print("  ❌ ERROR: Expected 2 NSP13 chains")

print("="*70)
```

## Section 3: Visualize Structure

```python
# Read PDB file
with open(PDB_FILE, 'r') as f:
    pdb_data = f.read()

# Create viewer
view = py3Dmol.view(width=800, height=600)
view.addModel(pdb_data, 'pdb')

# Color by chain
if len(nsp13_chains) >= 2:
    view.setStyle({'chain': nsp13_chains[0]}, {'cartoon': {'color': 'cyan', 'opacity': 0.8}})
    view.setStyle({'chain': nsp13_chains[1]}, {'cartoon': {'color': 'magenta', 'opacity': 0.8}})
    
    # Add labels
    nsp13_A = model[nsp13_chains[0]]
    residues_A = [r for r in nsp13_A.get_residues() if r.id[0] == ' ']
    if len(residues_A) > 10:
        mid_res_A = residues_A[len(residues_A)//2].id[1]
        view.addLabel(f"NSP13-A (Chain {nsp13_chains[0]})", 
                      {'fontColor': 'white', 'fontSize': 14, 'backgroundColor': 'cyan', 'backgroundOpacity': 0.8},
                      {'chain': nsp13_chains[0], 'resi': mid_res_A})
    
    nsp13_B = model[nsp13_chains[1]]
    residues_B = [r for r in nsp13_B.get_residues() if r.id[0] == ' ']
    if len(residues_B) > 10:
        mid_res_B = residues_B[len(residues_B)//2].id[1]
        view.addLabel(f"NSP13-B (Chain {nsp13_chains[1]})", 
                      {'fontColor': 'white', 'fontSize': 14, 'backgroundColor': 'magenta', 'backgroundOpacity': 0.8},
                      {'chain': nsp13_chains[1], 'resi': mid_res_B})

view.zoomTo()
view.setBackgroundColor('white')

print("3D Structure - NSP13 Helicase Homodimer:")
if len(nsp13_chains) >= 2:
    print(f"  Cyan = NSP13 Copy A (Chain {nsp13_chains[0]})")
    print(f"  Magenta = NSP13 Copy B (Chain {nsp13_chains[1]})")

view.show()
```

## Section 4: DISCOVER NSP13-NSP13 Hot Spots

```python
print("="*70)
print("DISCOVERING HOT SPOTS: NSP13-NSP13 HOMODIMER INTERFACE")
print("="*70)
print()

if len(nsp13_chains) < 2:
    print("⚠️  Missing required chains!")
    df_hot_nsp13 = pd.DataFrame()
else:
    nsp13_A = model[nsp13_chains[0]]
    nsp13_B = model[nsp13_chains[1]]
    
    # Get all atoms
    nsp13_A_atoms = [atom for res in nsp13_A if res.id[0] == ' ' for atom in res.get_atoms()]
    nsp13_B_atoms = [atom for res in nsp13_B if res.id[0] == ' ' for atom in res.get_atoms()]
    
    print(f"NSP13 Copy A (Chain {nsp13_chains[0]}): {len(nsp13_A_atoms)} atoms")
    print(f"NSP13 Copy B (Chain {nsp13_chains[1]}): {len(nsp13_B_atoms)} atoms")
    print()
    print("Searching for ALL charged interactions (< 5 Å)...")
    print("This may take a minute...")
    print()
    
    # Find ALL interface contacts < 5 Å
    interface_contacts = []
    for atom1 in nsp13_A_atoms:
        for atom2 in nsp13_B_atoms:
            dist = np.linalg.norm(atom1.get_coord() - atom2.get_coord())
            if dist < 5.0:
                res1 = atom1.get_parent()
                res2 = atom2.get_parent()
                interface_contacts.append((res1, res2, dist, atom1.get_name(), atom2.get_name()))
    
    print(f"✓ Found {len(interface_contacts)} atomic contacts < 5 Å")
    print()
    
    # Get unique residue pairs
    unique_pairs = {}
    for res1, res2, dist, atom1, atom2 in interface_contacts:
        key = (res1.id[1], res2.id[1])
        if key not in unique_pairs or dist < unique_pairs[key][2]:
            unique_pairs[key] = (res1, res2, dist, atom1, atom2)
    
    print(f"✓ Unique residue pairs: {len(unique_pairs)}")
    print()
    
    # DISCOVER ALL charged interactions
    print("="*70)
    print("CHARGED INTERACTION DISCOVERY")
    print("="*70)
    print()
    
    charged_positive = ['LYS', 'ARG']
    charged_negative = ['ASP', 'GLU']
    
    hot_spots_nsp13 = []
    for (pdb1, pdb2), (res1, res2, dist, atom1, atom2) in unique_pairs.items():
        resname1 = res1.get_resname()
        resname2 = res2.get_resname()
        
        # Check for charged interactions
        if (resname1 in charged_positive and resname2 in charged_negative) or \
           (resname1 in charged_negative and resname2 in charged_positive):
            
            interaction_type = 'Salt bridge' if dist < 4.0 else 'Ionic interaction'
            
            hot_spots_nsp13.append({
                'NSP13_A_Res': resname1,
                'NSP13_A_PDB': pdb1,
                'NSP13_B_Res': resname2,
                'NSP13_B_PDB': pdb2,
                'Distance': dist,
                'Atom1': atom1,
                'Atom2': atom2,
                'Type': interaction_type
            })
    
    if hot_spots_nsp13:
        df_hot_nsp13 = pd.DataFrame(hot_spots_nsp13).sort_values('Distance')
        
        print(f"✓ DISCOVERED {len(df_hot_nsp13)} charged interactions!")
        print()
        print("All Charged Interactions (sorted by distance):")
        print()
        print(df_hot_nsp13.to_string(index=False))
        print()
        
        # Highlight strongest
        strongest = df_hot_nsp13.iloc[0]
        print("="*70)
        print("🔥 STRONGEST HOT SPOT (NSP13-NSP13):")
        print(f"  {strongest['NSP13_A_Res']}{strongest['NSP13_A_PDB']} (Chain {nsp13_chains[0]}) ←→ {strongest['NSP13_B_Res']}{strongest['NSP13_B_PDB']} (Chain {nsp13_chains[1]})")
        print(f"  Distance: {strongest['Distance']:.2f} Å")
        print(f"  Type: {strongest['Type']}")
        print("="*70)
    else:
        print("⚠️  No charged interactions found < 5 Å")
        df_hot_nsp13 = pd.DataFrame()

print()
```

## Section 5: Visualize Hot Spot

```python
# Visualize hot spot
if not df_hot_nsp13.empty and len(nsp13_chains) >= 2:
    with open(PDB_FILE, 'r') as f:
        pdb_data = f.read()
    
    view = py3Dmol.view(width=800, height=600)
    view.addModel(pdb_data, 'pdb')
    
    # Show proteins transparent
    view.setStyle({'chain': nsp13_chains[0]}, {'cartoon': {'color': 'cyan', 'opacity': 0.3}})
    view.setStyle({'chain': nsp13_chains[1]}, {'cartoon': {'color': 'magenta', 'opacity': 0.3}})
    
    # Get hot spot residues
    top = df_hot_nsp13.iloc[0]
    
    # Convert to Python int (fix for numpy int64 issue)
    res_a_pdb = int(top['NSP13_A_PDB'])
    res_b_pdb = int(top['NSP13_B_PDB'])
    
    # Highlight as SPHERES
    view.addStyle({'chain': nsp13_chains[0], 'resi': res_a_pdb},
                  {'sphere': {'color': 'blue', 'radius': 1.0}})
    
    view.addStyle({'chain': nsp13_chains[1], 'resi': res_b_pdb},
                  {'sphere': {'color': 'red', 'radius': 1.0}})
    
    # Add labels with distance
    view.addLabel(f"{top['NSP13_A_Res']}{res_a_pdb}",
                  {'fontColor': 'white', 'fontSize': 16, 'backgroundColor': 'blue', 'backgroundOpacity': 0.8},
                  {'chain': nsp13_chains[0], 'resi': res_a_pdb})
    
    view.addLabel(f"{top['NSP13_B_Res']}{res_b_pdb} {top['Distance']:.1f} Å",
                  {'fontColor': 'white', 'fontSize': 16, 'backgroundColor': 'red', 'backgroundOpacity': 0.8},
                  {'chain': nsp13_chains[1], 'resi': res_b_pdb})
    
    # Center on hot spot
    view.zoomTo({'chain': nsp13_chains[0], 'resi': res_a_pdb})
    view.setBackgroundColor('white')
    
    print(f"Hot Spot: {top['NSP13_A_Res']}{res_a_pdb} ←→ {top['NSP13_B_Res']}{res_b_pdb}")
    print(f"Distance: {top['Distance']:.2f} Å")
    print()
    print("Blue spheres = NSP13 Chain A")
    print("Red spheres = NSP13 Chain B")
    
    view.show()
else:
    print("No hot spots to visualize")
```

## Section 6: Map NSP13-NSP13 Interface (10 Å)

```python
print("="*70)
print("MAPPING NSP13-NSP13 INTERFACE (10 Å cutoff)")
print("="*70)
print()

if not df_hot_nsp13.empty and len(nsp13_chains) >= 2:
    top_spot = df_hot_nsp13.iloc[0]
    res_pdb = int(top_spot['NSP13_A_PDB'])
    res_A = model[nsp13_chains[0]][res_pdb]
    center = res_A['CA'].get_coord()
    
    print(f"Center point: NSP13-A {top_spot['NSP13_A_Res']}{res_pdb}")
    print(f"Coordinates: ({center[0]:.1f}, {center[1]:.1f}, {center[2]:.1f})")
    print()
    print("Finding all residues within 10 Å...")
    print()
    
    interface_nsp13_full = []
    for chain_id in nsp13_chains:
        for res in model[chain_id]:
            if res.id[0] == ' ' and 'CA' in res:
                ca = res['CA'].get_coord()
                dist = np.linalg.norm(ca - center)
                
                if dist <= 10.0:
                    copy = 'NSP13-A' if chain_id == nsp13_chains[0] else 'NSP13-B'
                    interface_nsp13_full.append({
                        'Copy': copy,
                        'Chain': chain_id,
                        'Residue': res.get_resname(),
                        'PDB_Num': res.id[1],
                        'Distance': dist
                    })
    
    df_interface_nsp13 = pd.DataFrame(interface_nsp13_full).sort_values('Distance').reset_index(drop=True)
    
    print(f"✓ Found {len(df_interface_nsp13)} interface residues:")
    print()
    print(df_interface_nsp13.to_string(index=False))
    print()
    
    nsp13_A_count = len(df_interface_nsp13[df_interface_nsp13['Copy'] == 'NSP13-A'])
    nsp13_B_count = len(df_interface_nsp13[df_interface_nsp13['Copy'] == 'NSP13-B'])
    
    print("Interface composition:")
    print(f"  NSP13-A: {nsp13_A_count} residues")
    print(f"  NSP13-B: {nsp13_B_count} residues")
    print(f"  Total: {len(df_interface_nsp13)} residues")
else:
    print("⚠️  Cannot map interface")
    df_interface_nsp13 = pd.DataFrame()
    nsp13_A_count = nsp13_B_count = 0

print()
print("="*70)
```

## Section 7: Define Docking Grid Box

```python
print("="*70)
print("DOCKING GRID BOX PARAMETERS")
print("="*70)
print()

grid_box = {}

if not df_hot_nsp13.empty:
    top_spot = df_hot_nsp13.iloc[0]
    res_pdb = int(top_spot['NSP13_A_PDB'])
    res_A = model[nsp13_chains[0]][res_pdb]
    center = res_A['CA'].get_coord()
    
    grid_box = {
        'center_x': float(center[0]),
        'center_y': float(center[1]),
        'center_z': float(center[2]),
        'size_x': 25.0,
        'size_y': 25.0,
        'size_z': 25.0,
        'hot_spot': f"{top_spot['NSP13_A_Res']}{int(top_spot['NSP13_A_PDB'])}-{top_spot['NSP13_B_Res']}{int(top_spot['NSP13_B_PDB'])}"
    }
    
    print("INTERFACE: NSP13-NSP13 (Homodimer)")
    print(f"  Target: {grid_box['hot_spot']}")
    print(f"  Center: ({center[0]:.3f}, {center[1]:.3f}, {center[2]:.3f})")
    print(f"  Size: 25 × 25 × 25 Å³")
    print()
    print("="*70)
    print("FOR AUTODOCK VINA CONFIG FILE:")
    print("="*70)
    print(f"center_x = {grid_box['center_x']:.3f}")
    print(f"center_y = {grid_box['center_y']:.3f}")
    print(f"center_z = {grid_box['center_z']:.3f}")
    print(f"size_x = {grid_box['size_x']}")
    print(f"size_y = {grid_box['size_y']}")
    print(f"size_z = {grid_box['size_z']}")
else:
    print("⚠️  No hot spots found")

print()
print("="*70)
```

## Section 8: Summary Table

```python
# Build summary
summary_data = []

summary_data.append(['Property', 'Value'])
summary_data.append(['PDB ID', '6ZSL'])
summary_data.append(['Analysis Date', datetime.now().strftime('%Y-%m-%d')])
summary_data.append(['', ''])

summary_data.append(['STRUCTURE TYPE', 'Homodimer'])
summary_data.append(['Protein', 'NSP13 (Helicase)'])
if len(nsp13_chains) >= 2:
    summary_data.append(['Chain A', nsp13_chains[0]])
    summary_data.append(['Chain B', nsp13_chains[1]])
summary_data.append(['', ''])

summary_data.append(['HOT SPOTS', ''])
if not df_hot_nsp13.empty:
    top = df_hot_nsp13.iloc[0]
    summary_data.append(['Total Found', str(len(df_hot_nsp13))])
    summary_data.append(['Strongest', f"{top['NSP13_A_Res']}{int(top['NSP13_A_PDB'])}-{top['NSP13_B_Res']}{int(top['NSP13_B_PDB'])}"])
    summary_data.append(['Distance', f"{top['Distance']:.2f} Å"])
    summary_data.append(['Type', top['Type']])
else:
    summary_data.append(['Status', 'None found'])
summary_data.append(['', ''])

summary_data.append(['INTERFACE', ''])
if not df_interface_nsp13.empty:
    summary_data.append(['Total Residues', str(len(df_interface_nsp13))])
    summary_data.append(['NSP13-A Residues', str(nsp13_A_count)])
    summary_data.append(['NSP13-B Residues', str(nsp13_B_count)])
else:
    summary_data.append(['Status', 'Not mapped'])
summary_data.append(['', ''])

summary_data.append(['GRID BOX', ''])
if grid_box:
    summary_data.append(['Center', f"({grid_box['center_x']:.1f}, {grid_box['center_y']:.1f}, {grid_box['center_z']:.1f})"])
    summary_data.append(['Size', '25 × 25 × 25 Å³'])
else:
    summary_data.append(['Status', 'Not defined'])

df_summary = pd.DataFrame(summary_data[1:], columns=summary_data[0])

print("="*70)
print("6ZSL ANALYSIS SUMMARY")
print("NSP13 Helicase Homodimer")
print("="*70)
print()
print(df_summary.to_string(index=False))
print()
print("="*70)
```

## Section 9: Export Results

```python
print("Exporting results...")
print()

# Export hot spots
if not df_hot_nsp13.empty:
    csv_file = RESULTS_DIR / '6ZSL_NSP13-NSP13_hotspots.csv'
    df_hot_nsp13.to_csv(csv_file, index=False)
    print(f"✓ Hot spots: {csv_file}")

# Export interface
if not df_interface_nsp13.empty:
    csv_file = RESULTS_DIR / '6ZSL_NSP13-NSP13_interface.csv'
    df_interface_nsp13.to_csv(csv_file, index=False)
    print(f"✓ Interface: {csv_file}")

# Export JSON
analysis_data = {
    'pdb_id': '6ZSL',
    'analysis_date': datetime.now().strftime('%Y-%m-%d'),
    'structure_type': 'homodimer',
    'protein': 'NSP13_helicase',
    'chains': {'A': nsp13_chains[0], 'B': nsp13_chains[1]} if len(nsp13_chains) >= 2 else {},
    'interface': {}
}

if not df_hot_nsp13.empty:
    top = df_hot_nsp13.iloc[0]
    analysis_data['interface'] = {
        'hot_spots_count': len(df_hot_nsp13),
        'strongest_hot_spot': {
            'chain_A_residue': f"{top['NSP13_A_Res']}{int(top['NSP13_A_PDB'])}",
            'chain_B_residue': f"{top['NSP13_B_Res']}{int(top['NSP13_B_PDB'])}",
            'distance': float(top['Distance']),
            'type': top['Type']
        },
        'interface_residues': len(df_interface_nsp13) if not df_interface_nsp13.empty else 0,
        'grid_box': grid_box
    }

json_file = RESULTS_DIR / '6ZSL_analysis.json'
with open(json_file, 'w') as f:
    json.dump(analysis_data, f, indent=2)
print(f"✓ Analysis data: {json_file}")

# Export summary
summary_file = RESULTS_DIR / '6ZSL_summary.csv'
df_summary.to_csv(summary_file, index=False)
print(f"✓ Summary table: {summary_file}")

print()
print("✓ All results exported!")
print()
print(f"Results directory: {RESULTS_DIR.resolve()}")
```
