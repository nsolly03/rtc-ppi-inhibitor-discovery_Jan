## Section 1: Setup and Imports

```python
# Suppress warnings
import warnings
warnings.filterwarnings('ignore')

# Import libraries
import py3Dmol
from Bio import PDB
import numpy as np
import pandas as pd
import json
import os
from pathlib import Path
from datetime import datetime

# Configure paths (we're in Project root)
PDB_DIR = Path('data/structures/pdb')
RESULTS_DIR = Path('data/analysis_results')
PDB_FILE = PDB_DIR / '7DFG.pdb'

# Verify setup
RESULTS_DIR.mkdir(exist_ok=True, parents=True)
assert PDB_FILE.exists(), f"PDB file not found: {PDB_FILE}"

print("✓ Setup complete")
print(f"  PDB file: {PDB_FILE.resolve()}")
print(f"  Results: {RESULTS_DIR.resolve()}")
print(f"  File exists: {PDB_FILE.exists()}")
```

## Section 2: Load and Parse Structure

```python
# Parse structure
parser = PDB.PDBParser(QUIET=True)
structure = parser.get_structure('7DFG', str(PDB_FILE))
model = structure[0]

print("="*70)
print("STRUCTURE INFORMATION - 7DFG")
print("="*70)
print()

# AUTOMATIC chain identification by residue count
chain_assignments = {}
for chain in model:
    residues = [r for r in chain.get_residues() if r.id[0] == ' ']
    num_res = len(residues)
    
    if residues:
        first_res = residues[0].id[1]
        last_res = residues[-1].id[1]
    else:
        first_res = last_res = 'N/A'
    
    # Assign protein based on size
    if num_res > 700:
        protein = 'NSP12 (RdRp)'
        chain_assignments['nsp12'] = chain.id
    elif 70 < num_res < 90:
        protein = 'NSP7 (cofactor)'
        chain_assignments['nsp7'] = chain.id
    elif 180 < num_res < 210:
        protein = 'NSP8 (cofactor)'
        chain_assignments['nsp8'] = chain.id
    else:
        protein = 'Unknown'
    
    print(f"Chain {chain.id}: {num_res} residues (PDB {first_res}-{last_res}) → {protein}")

print()
print("Chain Assignments:")
print(f"  NSP12 = Chain {chain_assignments['nsp12']}")
print(f"  NSP7 = Chain {chain_assignments['nsp7']}")
print(f"  NSP8 = Chain {chain_assignments['nsp8']}")

print()
print("Target Interfaces for Discovery:")
print("  1. NSP12-NSP7 interface")
print("  2. NSP12-NSP8 interface")
print("="*70)
```

## Section 3: Visualize Structure (py3Dmol)

```python
# Read PDB file
with open(PDB_FILE, 'r') as f:
    pdb_data = f.read()

# Create viewer
view = py3Dmol.view(width=800, height=600)
view.addModel(pdb_data, 'pdb')

# Color by chain
view.setStyle({'chain': chain_assignments['nsp12']}, {'cartoon': {'color': 'cyan'}})
view.setStyle({'chain': chain_assignments['nsp7']}, {'cartoon': {'color': 'magenta'}})
view.setStyle({'chain': chain_assignments['nsp8']}, {'cartoon': {'color': 'yellow'}})

# Add labels
view.addLabel(f"NSP12 (Chain {chain_assignments['nsp12']})", 
              {'fontColor': 'white', 'backgroundColor': 'cyan', 'backgroundOpacity': 0.7},
              {'chain': chain_assignments['nsp12'], 'resi': 100})
view.addLabel(f"NSP7 (Chain {chain_assignments['nsp7']})", 
              {'fontColor': 'white', 'backgroundColor': 'magenta', 'backgroundOpacity': 0.7},
              {'chain': chain_assignments['nsp7'], 'resi': 10})
view.addLabel(f"NSP8 (Chain {chain_assignments['nsp8']})", 
              {'fontColor': 'white', 'backgroundColor': 'orange', 'backgroundOpacity': 0.7},
              {'chain': chain_assignments['nsp8'], 'resi': 50})

view.zoomTo()

print("3D Structure - RdRp Holoenzyme:")
print(f"  Cyan = NSP12 (Chain {chain_assignments['nsp12']})")
print(f"  Magenta = NSP7 (Chain {chain_assignments['nsp7']})")
print(f"  Yellow = NSP8 (Chain {chain_assignments['nsp8']})")

view.show()
```
