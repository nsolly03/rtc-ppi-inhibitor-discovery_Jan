## Section 1: Setup

```python
import warnings
warnings.filterwarnings('ignore')

import py3Dmol
from Bio import PDB
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import json

PDB_DIR = Path('../data/structures/pdb')
RESULTS_DIR = Path('../data/analysis_results')
PDB_FILE = PDB_DIR / '7DFG.pdb'

assert PDB_FILE.exists(), f'PDB file not found: {PDB_FILE}'
RESULTS_DIR.mkdir(exist_ok=True, parents=True)

print('='*70)
print('7DFG COMPREHENSIVE VALIDATION')
print('='*70)
print()
print('Target: NSP12-NSP7-NSP8 RdRp Complex')
print('Interfaces: 2 (NSP12-NSP7 and NSP12-NSP8)')
print('Hot spots to validate: 3')
print()
print('='*70)
```

## Section 2: Load Structure and Identify Chains

```python
parser = PDB.PDBParser(QUIET=True)
structure = parser.get_structure('7DFG', str(PDB_FILE))
model = structure[0]

print('='*70)
print('STRUCTURE INFORMATION')
print('='*70)
print()

# Analyze chains
chain_info = {}
for chain in model:
    residues = [r for r in chain.get_residues() if r.id[0] == ' ']
    if residues and len(residues) > 30:
        first_res = residues[0].get_resname()
        if first_res not in ['A', 'U', 'G', 'C', 'DA', 'DT', 'DG', 'DC']:
            chain_info[chain.id] = {
                'num_res': len(residues),
                'first': residues[0].id[1],
                'last': residues[-1].id[1]
            }
            print(f'Chain {chain.id}: {len(residues)} residues (PROTEIN)')

print()

# Identify proteins by size
# NSP12 (RdRp): ~932 residues (largest)
# NSP7: ~83 residues
# NSP8: ~198 residues

nsp12_chain = None
nsp7_chain = None
nsp8_chain = None

for chain_id, info in chain_info.items():
    if info['num_res'] > 500:
        nsp12_chain = chain_id
        print(f'NSP12 (RdRp) = Chain {chain_id} ({info["num_res"]} residues)')
    elif info['num_res'] < 100:
        nsp7_chain = chain_id
        print(f'NSP7 = Chain {chain_id} ({info["num_res"]} residues)')
    else:
        nsp8_chain = chain_id
        print(f'NSP8 = Chain {chain_id} ({info["num_res"]} residues)')

print()
print('Target Interfaces:')
print('  1. NSP12-NSP7')
print('  2. NSP12-NSP8')
print('='*70)
```

## Section 3: Interface 1 - NSP12-NSP7 Validation

```python
print('='*70)
print('INTERFACE 1: NSP12-NSP7')
print('='*70)
print()
print('Extracting charged residues...')
print()

charged_positive = ['LYS', 'ARG', 'HIS']
charged_negative = ['ASP', 'GLU']

# Extract from NSP12
nsp12 = model[nsp12_chain]
nsp12_charged = []

for residue in nsp12:
    if residue.id[0] == ' ':
        resname = residue.get_resname()
        if resname in charged_positive + charged_negative:
            nsp12_charged.append({
                'chain': nsp12_chain,
                'resname': resname,
                'resnum': residue.id[1],
                'residue': residue,
                'type': 'positive' if resname in charged_positive else 'negative'
            })

# Extract from NSP7
nsp7 = model[nsp7_chain]
nsp7_charged = []

for residue in nsp7:
    if residue.id[0] == ' ':
        resname = residue.get_resname()
        if resname in charged_positive + charged_negative:
            nsp7_charged.append({
                'chain': nsp7_chain,
                'resname': resname,
                'resnum': residue.id[1],
                'residue': residue,
                'type': 'positive' if resname in charged_positive else 'negative'
            })

print(f'NSP12 (Chain {nsp12_chain}): {len(nsp12_charged)} charged residues')
print(f'NSP7 (Chain {nsp7_chain}): {len(nsp7_charged)} charged residues')
print()

# Calculate distances
sidechain_atoms = {
    'LYS': ['NZ'],
    'ARG': ['NH1', 'NH2'],
    'HIS': ['ND1', 'NE2'],
    'ASP': ['OD1', 'OD2'],
    'GLU': ['OE1', 'OE2']
}

def get_sidechain_distance(res1, res2):
    resname1 = res1.get_resname()
    resname2 = res2.get_resname()
    atoms1 = sidechain_atoms.get(resname1, [])
    atoms2 = sidechain_atoms.get(resname2, [])
    min_dist = float('inf')
    min_atoms = (None, None)
    for atom1_name in atoms1:
        if atom1_name not in res1:
            continue
        atom1 = res1[atom1_name]
        for atom2_name in atoms2:
            if atom2_name not in res2:
                continue
            atom2 = res2[atom2_name]
            dist = np.linalg.norm(atom1.get_coord() - atom2.get_coord())
            if dist < min_dist:
                min_dist = dist
                min_atoms = (atom1_name, atom2_name)
    return min_dist, min_atoms[0], min_atoms[1]

nsp12_nsp7_interactions = []

for r1_info in nsp12_charged:
    for r2_info in nsp7_charged:
        if r1_info['type'] != r2_info['type']:
            res1 = r1_info['residue']
            res2 = r2_info['residue']
            dist, atom1, atom2 = get_sidechain_distance(res1, res2)
            if dist != float('inf'):
                nsp12_nsp7_interactions.append({
                    'NSP12_Res': r1_info['resname'],
                    'NSP12_Num': r1_info['resnum'],
                    'NSP12_Atom': atom1,
                    'NSP7_Res': r2_info['resname'],
                    'NSP7_Num': r2_info['resnum'],
                    'NSP7_Atom': atom2,
                    'Distance': dist,
                    'Type': 'Salt bridge' if dist < 4.0 else 'Ionic'
                })

df_nsp12_nsp7 = pd.DataFrame(nsp12_nsp7_interactions).sort_values('Distance').reset_index(drop=True)

print(f'Calculated {len(df_nsp12_nsp7)} NSP12-NSP7 interactions')
print()
print('TOP 5 NSP12-NSP7 INTERACTIONS:')
print(df_nsp12_nsp7.head(5).to_string(index=False))
print()

# Validate GLU431-LYS2
glu431_lys2 = df_nsp12_nsp7[
    ((df_nsp12_nsp7['NSP12_Res'] == 'GLU') & (df_nsp12_nsp7['NSP12_Num'] == 431) & 
     (df_nsp12_nsp7['NSP7_Res'] == 'LYS') & (df_nsp12_nsp7['NSP7_Num'] == 2))
]

if not glu431_lys2.empty:
    row = glu431_lys2.iloc[0]
    rank = df_nsp12_nsp7.index[df_nsp12_nsp7['Distance'] == row['Distance']].tolist()[0] + 1
    print('GLU431-LYS2 VALIDATION:')
    print(f'  Previous: 3.29 Angstrom')
    print(f'  Current: {row["Distance"]:.2f} Angstrom')
    print(f'  Rank: #{rank}')
    print(f'  Status: {"CONFIRMED" if rank == 1 else "Check"}')
else:
    print('GLU431-LYS2 not found - checking numbering...')

print()
print('='*70)
```

## Section 4: Interface 2 - NSP12-NSP8 Validation

```python
print('='*70)
print('INTERFACE 2: NSP12-NSP8')
print('='*70)
print()

# Extract from NSP8
nsp8 = model[nsp8_chain]
nsp8_charged = []

for residue in nsp8:
    if residue.id[0] == ' ':
        resname = residue.get_resname()
        if resname in charged_positive + charged_negative:
            nsp8_charged.append({
                'chain': nsp8_chain,
                'resname': resname,
                'resnum': residue.id[1],
                'residue': residue,
                'type': 'positive' if resname in charged_positive else 'negative'
            })

print(f'NSP12 (Chain {nsp12_chain}): {len(nsp12_charged)} charged residues')
print(f'NSP8 (Chain {nsp8_chain}): {len(nsp8_charged)} charged residues')
print()

# Calculate distances
nsp12_nsp8_interactions = []

for r1_info in nsp12_charged:
    for r2_info in nsp8_charged:
        if r1_info['type'] != r2_info['type']:
            res1 = r1_info['residue']
            res2 = r2_info['residue']
            dist, atom1, atom2 = get_sidechain_distance(res1, res2)
            if dist != float('inf'):
                nsp12_nsp8_interactions.append({
                    'NSP12_Res': r1_info['resname'],
                    'NSP12_Num': r1_info['resnum'],
                    'NSP12_Atom': atom1,
                    'NSP8_Res': r2_info['resname'],
                    'NSP8_Num': r2_info['resnum'],
                    'NSP8_Atom': atom2,
                    'Distance': dist,
                    'Type': 'Salt bridge' if dist < 4.0 else 'Ionic'
                })

df_nsp12_nsp8 = pd.DataFrame(nsp12_nsp8_interactions).sort_values('Distance').reset_index(drop=True)

print(f'Calculated {len(df_nsp12_nsp8)} NSP12-NSP8 interactions')
print()
print('TOP 5 NSP12-NSP8 INTERACTIONS:')
print(df_nsp12_nsp8.head(5).to_string(index=False))
print()

# Validate ARG331-ASP112
arg331_asp112 = df_nsp12_nsp8[
    ((df_nsp12_nsp8['NSP12_Res'] == 'ARG') & (df_nsp12_nsp8['NSP12_Num'] == 331) & 
     (df_nsp12_nsp8['NSP8_Res'] == 'ASP') & (df_nsp12_nsp8['NSP8_Num'] == 112))
]

if not arg331_asp112.empty:
    row = arg331_asp112.iloc[0]
    rank = df_nsp12_nsp8.index[df_nsp12_nsp8['Distance'] == row['Distance']].tolist()[0] + 1
    print('ARG331-ASP112 VALIDATION:')
    print(f'  Previous: 3.02 Angstrom')
    print(f'  Current: {row["Distance"]:.2f} Angstrom')
    print(f'  Rank: #{rank}')
    print(f'  Status: {"CONFIRMED" if rank == 1 else "Check"}')
else:
    print('ARG331-ASP112 not found')

print()

# Validate ASP863-LYS58
asp863_lys58 = df_nsp12_nsp8[
    ((df_nsp12_nsp8['NSP12_Res'] == 'ASP') & (df_nsp12_nsp8['NSP12_Num'] == 863) & 
     (df_nsp12_nsp8['NSP8_Res'] == 'LYS') & (df_nsp12_nsp8['NSP8_Num'] == 58))
]

if not asp863_lys58.empty:
    row = asp863_lys58.iloc[0]
    rank = df_nsp12_nsp8.index[df_nsp12_nsp8['Distance'] == row['Distance']].tolist()[0] + 1
    print('ASP863-LYS58 VALIDATION:')
    print(f'  Previous: 3.66 Angstrom')
    print(f'  Current: {row["Distance"]:.2f} Angstrom')
    print(f'  Rank: #{rank}')
    print(f'  Status: {"CONFIRMED" if rank <= 2 else "Check"}')
else:
    print('ASP863-LYS58 not found')

print()
print('='*70)
```

## Section 5: Export Results

```python
# Export Interface 1 results
df_nsp12_nsp7.to_csv(RESULTS_DIR / '7DFG_NSP12_NSP7_validation.csv', index=False)
print('Exported NSP12-NSP7 results')

# Export Interface 2 results
df_nsp12_nsp8.to_csv(RESULTS_DIR / '7DFG_NSP12_NSP8_validation.csv', index=False)
print('Exported NSP12-NSP8 results')

# Summary
summary = {
    'pdb_id': '7DFG',
    'interfaces': 2,
    'nsp12_nsp7': {
        'interactions': len(df_nsp12_nsp7),
        'strongest': {
            'residues': f"{df_nsp12_nsp7.iloc[0]['NSP12_Res']}{df_nsp12_nsp7.iloc[0]['NSP12_Num']}-{df_nsp12_nsp7.iloc[0]['NSP7_Res']}{df_nsp12_nsp7.iloc[0]['NSP7_Num']}",
            'distance': float(df_nsp12_nsp7.iloc[0]['Distance'])
        }
    },
    'nsp12_nsp8': {
        'interactions': len(df_nsp12_nsp8),
        'strongest': {
            'residues': f"{df_nsp12_nsp8.iloc[0]['NSP12_Res']}{df_nsp12_nsp8.iloc[0]['NSP12_Num']}-{df_nsp12_nsp8.iloc[0]['NSP8_Res']}{df_nsp12_nsp8.iloc[0]['NSP8_Num']}",
            'distance': float(df_nsp12_nsp8.iloc[0]['Distance'])
        }
    }
}

with open(RESULTS_DIR / '7DFG_validation_summary.json', 'w') as f:
    json.dump(summary, f, indent=2)

print('Exported summary')
print()
print('7DFG VALIDATION COMPLETE')
```
