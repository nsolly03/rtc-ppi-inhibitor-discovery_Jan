## Section 1: Setup

```python
import warnings
warnings.filterwarnings('ignore')

import py3Dmol
from Bio import PDB
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import json

PDB_DIR = Path('../data/structures/pdb')
RESULTS_DIR = Path('../data/analysis_results')
PDB_FILE = PDB_DIR / '6ZSL.pdb'

assert PDB_FILE.exists(), f'PDB file not found: {PDB_FILE}'
RESULTS_DIR.mkdir(exist_ok=True, parents=True)

print('='*70)
print('6ZSL COMPREHENSIVE VALIDATION')
print('='*70)
print()
print('Target: NSP13-NSP13 Helicase Homodimer')
print('Goal: Validate homodimer hot spots')
print('Challenge: Handle symmetric vs asymmetric interactions')
print()
print('='*70)
```

## Section 2: Load Structure and Identify Chains

```python
parser = PDB.PDBParser(QUIET=True)
structure = parser.get_structure('6ZSL', str(PDB_FILE))
model = structure[0]

print('='*70)
print('STRUCTURE INFORMATION')
print('='*70)
print()

# Analyze chains
for chain in model:
    residues = [r for r in chain.get_residues() if r.id[0] == ' ']
    if residues:
        print(f'Chain {chain.id}: {len(residues)} residues')
        print(f'  Range: {residues[0].id[1]} to {residues[-1].id[1]}')

print()

# For homodimer, both chains should be similar size
chains = [c.id for c in model.get_chains()]
print(f'Total chains: {len(chains)}')
print(f'Chain IDs: {chains}')
print()
print('Homodimer: Both chains are NSP13 (helicase)')
print(f'  Copy 1: Chain {chains[0]}')
print(f'  Copy 2: Chain {chains[1]}')

nsp13_chain1 = chains[0]
nsp13_chain2 = chains[1]

print()
print('Target Interface: NSP13-NSP13 (homodimer)')
print('='*70)
```

## Section 3: Extract ALL Charged Residues

```python
print('='*70)
print('EXTRACTING ALL CHARGED RESIDUES')
print('='*70)
print()

charged_positive = ['LYS', 'ARG', 'HIS']
charged_negative = ['ASP', 'GLU']

# Extract from Chain 1
nsp13_1 = model[nsp13_chain1]
nsp13_1_charged = []

for residue in nsp13_1:
    if residue.id[0] == ' ':
        resname = residue.get_resname()
        if resname in charged_positive + charged_negative:
            nsp13_1_charged.append({
                'chain': nsp13_chain1,
                'resname': resname,
                'resnum': residue.id[1],
                'residue': residue,
                'type': 'positive' if resname in charged_positive else 'negative'
            })

# Extract from Chain 2
nsp13_2 = model[nsp13_chain2]
nsp13_2_charged = []

for residue in nsp13_2:
    if residue.id[0] == ' ':
        resname = residue.get_resname()
        if resname in charged_positive + charged_negative:
            nsp13_2_charged.append({
                'chain': nsp13_chain2,
                'resname': resname,
                'resnum': residue.id[1],
                'residue': residue,
                'type': 'positive' if resname in charged_positive else 'negative'
            })

print(f'NSP13 Copy 1 (Chain {nsp13_chain1}):')
print(f'  Total charged: {len(nsp13_1_charged)}')
print(f'  Positive: {len([r for r in nsp13_1_charged if r["type"] == "positive"])}')
print(f'  Negative: {len([r for r in nsp13_1_charged if r["type"] == "negative"])}')
print()

print(f'NSP13 Copy 2 (Chain {nsp13_chain2}):')
print(f'  Total charged: {len(nsp13_2_charged)}')
print(f'  Positive: {len([r for r in nsp13_2_charged if r["type"] == "positive"])}')
print(f'  Negative: {len([r for r in nsp13_2_charged if r["type"] == "negative"])}')
print()

# Check for specific residues
print('CHECKING FOR EXPECTED HOT SPOTS:')
for res_num in [390, 365, 189, 341]:
    found_1 = any(r['resnum'] == res_num for r in nsp13_1_charged)
    found_2 = any(r['resnum'] == res_num for r in nsp13_2_charged)
    if found_1:
        res_info = [r for r in nsp13_1_charged if r['resnum'] == res_num][0]
        print(f'  Found {res_info["resname"]}{res_num} in Chain {nsp13_chain1}')
    if found_2:
        res_info = [r for r in nsp13_2_charged if r['resnum'] == res_num][0]
        print(f'  Found {res_info["resname"]}{res_num} in Chain {nsp13_chain2}')

print()
print(f'Total pairwise comparisons: {len(nsp13_1_charged) * len(nsp13_2_charged)}')
print('='*70)
```

## Section 4: Calculate Sidechain Distances

```python
print('='*70)
print('CALCULATING SIDECHAIN-BASED DISTANCES')
print('='*70)
print()

sidechain_atoms = {
    'LYS': ['NZ'],
    'ARG': ['NH1', 'NH2'],
    'HIS': ['ND1', 'NE2'],
    'ASP': ['OD1', 'OD2'],
    'GLU': ['OE1', 'OE2']
}

def get_sidechain_distance(res1, res2):
    resname1 = res1.get_resname()
    resname2 = res2.get_resname()
    
    atoms1 = sidechain_atoms.get(resname1, [])
    atoms2 = sidechain_atoms.get(resname2, [])
    
    min_dist = float('inf')
    min_atoms = (None, None)
    
    for atom1_name in atoms1:
        if atom1_name not in res1:
            continue
        atom1 = res1[atom1_name]
        
        for atom2_name in atoms2:
            if atom2_name not in res2:
                continue
            atom2 = res2[atom2_name]
            
            dist = np.linalg.norm(atom1.get_coord() - atom2.get_coord())
            if dist < min_dist:
                min_dist = dist
                min_atoms = (atom1_name, atom2_name)
    
    return min_dist, min_atoms[0], min_atoms[1]

print('Computing all pairwise sidechain distances...')
print('(Homodimer: Chain 1 vs Chain 2)')
print()

all_interactions = []

for r1_info in nsp13_1_charged:
    for r2_info in nsp13_2_charged:
        if r1_info['type'] != r2_info['type']:
            res1 = r1_info['residue']
            res2 = r2_info['residue']
            
            dist, atom1, atom2 = get_sidechain_distance(res1, res2)
            
            if dist != float('inf'):
                all_interactions.append({
                    'Chain1_Res': r1_info['resname'],
                    'Chain1_Num': r1_info['resnum'],
                    'Chain1_Atom': atom1,
                    'Chain2_Res': r2_info['resname'],
                    'Chain2_Num': r2_info['resnum'],
                    'Chain2_Atom': atom2,
                    'Distance': dist,
                    'Type': 'Salt bridge' if dist < 4.0 else 'Ionic'
                })

df_all = pd.DataFrame(all_interactions).sort_values('Distance').reset_index(drop=True)

print(f'Calculated {len(df_all)} pairwise distances')
print()
print('='*70)
```

## Section 5: Analyze Top Interactions

```python
print('='*70)
print('TOP 10 INTERACTIONS (Sidechain-Based)')
print('='*70)
print()

print(df_all.head(10).to_string(index=False))
print()
print('='*70)
print()

# Look for ARG390-GLU365
arg390_glu365 = df_all[
    ((df_all['Chain1_Res'] == 'ARG') & (df_all['Chain1_Num'] == 390) & 
     (df_all['Chain2_Res'] == 'GLU') & (df_all['Chain2_Num'] == 365)) |
    ((df_all['Chain2_Res'] == 'ARG') & (df_all['Chain2_Num'] == 390) & 
     (df_all['Chain1_Res'] == 'GLU') & (df_all['Chain1_Num'] == 365))
]

print('HOT SPOT 1 - ARG390-GLU365 VALIDATION:')
print()

if not arg390_glu365.empty:
    row = arg390_glu365.iloc[0]
    rank = df_all.index[df_all['Distance'] == row['Distance']].tolist()[0] + 1
    
    print(f'  Previous (Week 3): 3.05 Angstrom')
    print(f'  Current (Validated): {row["Distance"]:.2f} Angstrom')
    print(f'  Atoms: {row["Chain1_Atom"]} - {row["Chain2_Atom"]}')
    print(f'  Rank: #{rank} out of {len(df_all)}')
    print(f'  Type: {row["Type"]}')
    print()
    
    if rank == 1:
        print('  CONFIRMED: ARG390-GLU365 is #1 hot spot!')
    else:
        print(f'  ARG390-GLU365 is #{rank}')
else:
    print('  ARG390-GLU365 not found')

print()
print('='*70)
print()

# Look for LYS189-GLU341
lys189_glu341 = df_all[
    ((df_all['Chain1_Res'] == 'LYS') & (df_all['Chain1_Num'] == 189) & 
     (df_all['Chain2_Res'] == 'GLU') & (df_all['Chain2_Num'] == 341)) |
    ((df_all['Chain2_Res'] == 'LYS') & (df_all['Chain2_Num'] == 189) & 
     (df_all['Chain1_Res'] == 'GLU') & (df_all['Chain1_Num'] == 341))
]

print('HOT SPOT 2 - LYS189-GLU341 VALIDATION:')
print()

if not lys189_glu341.empty:
    row = lys189_glu341.iloc[0]
    rank = df_all.index[df_all['Distance'] == row['Distance']].tolist()[0] + 1
    
    print(f'  Previous (Week 3): 3.80 Angstrom')
    print(f'  Current (Validated): {row["Distance"]:.2f} Angstrom')
    print(f'  Atoms: {row["Chain1_Atom"]} - {row["Chain2_Atom"]}')
    print(f'  Rank: #{rank} out of {len(df_all)}')
    print(f'  Type: {row["Type"]}')
    print()
    
    if rank == 2:
        print('  CONFIRMED: LYS189-GLU341 is #2 hot spot!')
    else:
        print(f'  LYS189-GLU341 is #{rank}')
else:
    print('  LYS189-GLU341 not found')

print()
print('='*70)
```

## Section 6: Distance Matrix Heatmap

```python
print('Creating distance matrix heatmap...')
print()

df_all['Chain1_Label'] = df_all['Chain1_Res'] + df_all['Chain1_Num'].astype(str)
df_all['Chain2_Label'] = df_all['Chain2_Res'] + df_all['Chain2_Num'].astype(str)

df_close = df_all[df_all['Distance'] < 10.0].copy()

if len(df_close) > 0:
    matrix = df_close.pivot_table(
        index='Chain1_Label',
        columns='Chain2_Label',
        values='Distance',
        aggfunc='min'
    )
    
    plt.figure(figsize=(14, 12))
    sns.heatmap(
        matrix,
        cmap='RdYlGn_r',
        vmin=0,
        vmax=10,
        cbar_kws={'label': 'Distance (Angstrom)'},
        square=True
    )
    plt.title('6ZSL: NSP13-NSP13 Homodimer Distance Matrix', fontsize=14, fontweight='bold')
    plt.xlabel('NSP13 Copy 2', fontsize=12)
    plt.ylabel('NSP13 Copy 1', fontsize=12)
    plt.tight_layout()
    plt.savefig(RESULTS_DIR / '6ZSL_distance_matrix_comprehensive.png', dpi=300, bbox_inches='tight')
    plt.show()
    
    print('Heatmap saved')
else:
    print('No interactions < 10A to plot')
```

## Section 7: Visualize Top Hot Spots

```python
if not df_all.empty:
    print('='*70)
    print('VISUALIZING TOP 2 HOT SPOTS')
    print('='*70)
    print()
    
    with open(PDB_FILE, 'r') as f:
        pdb_data = f.read()
    
    view = py3Dmol.view(width=900, height=700)
    view.addModel(pdb_data, 'pdb')
    
    # Transparent proteins
    view.setStyle({'chain': nsp13_chain1}, {'cartoon': {'color': 'red', 'opacity': 0.3}})
    view.setStyle({'chain': nsp13_chain2}, {'cartoon': {'color': 'orange', 'opacity': 0.3}})
    
    # Show top 2 hot spots
    for i in range(min(2, len(df_all))):
        row = df_all.iloc[i]
        color1 = 'blue' if i == 0 else 'cyan'
        color2 = 'green' if i == 0 else 'yellow'
        
        view.addStyle({'chain': nsp13_chain1, 'resi': int(row['Chain1_Num'])},
                      {'sphere': {'color': color1, 'radius': 1.0}})
        view.addStyle({'chain': nsp13_chain2, 'resi': int(row['Chain2_Num'])},
                      {'sphere': {'color': color2, 'radius': 1.0}})
        
        view.addLabel(f"#{i+1}: {row['Chain1_Res']}{int(row['Chain1_Num'])}",
                      {'fontColor': 'white', 'fontSize': 14, 'backgroundColor': color1, 'backgroundOpacity': 0.8},
                      {'chain': nsp13_chain1, 'resi': int(row['Chain1_Num'])})
        
        view.addLabel(f"{row['Chain2_Res']}{int(row['Chain2_Num'])} {row['Distance']:.2f}A",
                      {'fontColor': 'white', 'fontSize': 14, 'backgroundColor': color2, 'backgroundOpacity': 0.8},
                      {'chain': nsp13_chain2, 'resi': int(row['Chain2_Num'])})
    
    view.zoomTo()
    view.setBackgroundColor('white')
    
    print('Showing top 2 hot spots')
    print('  Red/Orange = NSP13 copies')
    print('  Blue/Green = Hot spots')
    
    view.show()
```

## Section 8: Export Results

```python
print('='*70)
print('EXPORTING RESULTS')
print('='*70)
print()

csv_file = RESULTS_DIR / '6ZSL_comprehensive_validation_all.csv'
df_all.to_csv(csv_file, index=False)
print(f'All interactions: {csv_file}')

top20_file = RESULTS_DIR / '6ZSL_comprehensive_validation_top20.csv'
df_all.head(20).to_csv(top20_file, index=False)
print(f'Top 20: {top20_file}')

summary_data = {
    'pdb_id': '6ZSL',
    'analysis_type': 'comprehensive_validation',
    'method': 'sidechain_based',
    'structure_type': 'homodimer',
    'total_interactions': len(df_all),
    'strongest': {
        'residues': f"{df_all.iloc[0]['Chain1_Res']}{df_all.iloc[0]['Chain1_Num']}-{df_all.iloc[0]['Chain2_Res']}{df_all.iloc[0]['Chain2_Num']}",
        'distance': float(df_all.iloc[0]['Distance']),
        'atoms': f"{df_all.iloc[0]['Chain1_Atom']}-{df_all.iloc[0]['Chain2_Atom']}"
    }
}

json_file = RESULTS_DIR / '6ZSL_comprehensive_validation.json'
with open(json_file, 'w') as f:
    json.dump(summary_data, f, indent=2)
print(f'Summary: {json_file}')

print()
print('All results exported!')
print('='*70)
```
