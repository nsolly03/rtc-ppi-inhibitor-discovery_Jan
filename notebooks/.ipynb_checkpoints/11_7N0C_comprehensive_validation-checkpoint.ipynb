## Section 1: Setup

```python
import warnings
warnings.filterwarnings('ignore')

import py3Dmol
from Bio import PDB
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import json

PDB_DIR = Path('../data/structures/pdb')
RESULTS_DIR = Path('../data/analysis_results')
PDB_FILE = PDB_DIR / '7N0C.pdb'

assert PDB_FILE.exists(), f'PDB file not found: {PDB_FILE}'
RESULTS_DIR.mkdir(exist_ok=True, parents=True)

print('='*70)
print('7N0C COMPREHENSIVE VALIDATION')
print('='*70)
print()
print('Target: NSP10-NSP14 Exonuclease Complex')
print('Goal: Validate K93-D126 as #1 overall')
print('Special focus: K76 conservation check')
print()
print('='*70)
```

## Section 2: Load Structure and Identify Chains

```python
parser = PDB.PDBParser(QUIET=True)
structure = parser.get_structure('7N0C', str(PDB_FILE))
model = structure[0]

print('='*70)
print('STRUCTURE INFORMATION')
print('='*70)
print()

# Analyze all chains
chain_info = {}
for chain in model:
    residues = [r for r in chain.get_residues() if r.id[0] == ' ']
    if residues:
        chain_info[chain.id] = {
            'num_res': len(residues),
            'first': residues[0].id[1],
            'last': residues[-1].id[1]
        }
        print(f'Chain {chain.id}: {len(residues)} residues')
        print(f'  Range: {residues[0].id[1]} to {residues[-1].id[1]}')

print()

# Identify NSP10 and NSP14 based on size
# NSP10 ~139 residues (smaller)
# NSP14 ~527 residues (larger)

nsp10_chain = None
nsp14_chain = None

for chain_id, info in chain_info.items():
    if info['num_res'] < 200:
        nsp10_chain = chain_id
        print(f'NSP10 = Chain {chain_id} ({info["num_res"]} residues)')
    else:
        nsp14_chain = chain_id
        print(f'NSP14 = Chain {chain_id} ({info["num_res"]} residues)')

print()
print('Target Interface: NSP10-NSP14')
print('='*70)
```

## Section 3: Extract ALL Charged Residues

```python
print('='*70)
print('EXTRACTING ALL CHARGED RESIDUES')
print('='*70)
print()

charged_positive = ['LYS', 'ARG', 'HIS']
charged_negative = ['ASP', 'GLU']

# Extract from NSP10
nsp10 = model[nsp10_chain]
nsp10_charged = []

for residue in nsp10:
    if residue.id[0] == ' ':
        resname = residue.get_resname()
        if resname in charged_positive + charged_negative:
            nsp10_charged.append({
                'chain': nsp10_chain,
                'resname': resname,
                'resnum': residue.id[1],
                'residue': residue,
                'type': 'positive' if resname in charged_positive else 'negative'
            })

# Extract from NSP14
nsp14 = model[nsp14_chain]
nsp14_charged = []

for residue in nsp14:
    if residue.id[0] == ' ':
        resname = residue.get_resname()
        if resname in charged_positive + charged_negative:
            nsp14_charged.append({
                'chain': nsp14_chain,
                'resname': resname,
                'resnum': residue.id[1],
                'residue': residue,
                'type': 'positive' if resname in charged_positive else 'negative'
            })

print(f'NSP10 (Chain {nsp10_chain}):')
print(f'  Total charged: {len(nsp10_charged)}')
print(f'  Positive: {len([r for r in nsp10_charged if r["type"] == "positive"])}')
print(f'  Negative: {len([r for r in nsp10_charged if r["type"] == "negative"])}')
print()

print(f'NSP14 (Chain {nsp14_chain}):')
print(f'  Total charged: {len(nsp14_charged)}')
print(f'  Positive: {len([r for r in nsp14_charged if r["type"] == "positive"])}')
print(f'  Negative: {len([r for r in nsp14_charged if r["type"] == "negative"])}')
print()

# Check for K76 and K93
print('SPECIAL CHECK - K76 and K93 in NSP10:')
k76_found = False
k93_found = False

for r in nsp10_charged:
    if r['resname'] == 'LYS' and r['resnum'] == 76:
        print(f'  Found K76 in Chain {r["chain"]}')
        k76_found = True
    if r['resname'] == 'LYS' and r['resnum'] == 93:
        print(f'  Found K93 in Chain {r["chain"]}')
        k93_found = True

if not k76_found:
    print('  K76 not found (may use different numbering)')
if not k93_found:
    print('  K93 not found (may use different numbering)')

print()
print(f'Total pairwise comparisons: {len(nsp10_charged) * len(nsp14_charged)}')
print('='*70)
```

## Section 4: Calculate Sidechain Distances

```python
print('='*70)
print('CALCULATING SIDECHAIN-BASED DISTANCES')
print('='*70)
print()

sidechain_atoms = {
    'LYS': ['NZ'],
    'ARG': ['NH1', 'NH2'],
    'HIS': ['ND1', 'NE2'],
    'ASP': ['OD1', 'OD2'],
    'GLU': ['OE1', 'OE2']
}

def get_sidechain_distance(res1, res2):
    resname1 = res1.get_resname()
    resname2 = res2.get_resname()
    
    atoms1 = sidechain_atoms.get(resname1, [])
    atoms2 = sidechain_atoms.get(resname2, [])
    
    min_dist = float('inf')
    min_atoms = (None, None)
    
    for atom1_name in atoms1:
        if atom1_name not in res1:
            continue
        atom1 = res1[atom1_name]
        
        for atom2_name in atoms2:
            if atom2_name not in res2:
                continue
            atom2 = res2[atom2_name]
            
            dist = np.linalg.norm(atom1.get_coord() - atom2.get_coord())
            if dist < min_dist:
                min_dist = dist
                min_atoms = (atom1_name, atom2_name)
    
    return min_dist, min_atoms[0], min_atoms[1]

print('Computing all pairwise sidechain distances...')
print()

all_interactions = []

for r1_info in nsp10_charged:
    for r2_info in nsp14_charged:
        if r1_info['type'] != r2_info['type']:
            res1 = r1_info['residue']
            res2 = r2_info['residue']
            
            dist, atom1, atom2 = get_sidechain_distance(res1, res2)
            
            if dist != float('inf'):
                all_interactions.append({
                    'NSP10_Res': r1_info['resname'],
                    'NSP10_Num': r1_info['resnum'],
                    'NSP10_Atom': atom1,
                    'NSP14_Res': r2_info['resname'],
                    'NSP14_Num': r2_info['resnum'],
                    'NSP14_Atom': atom2,
                    'Distance': dist,
                    'Type': 'Salt bridge' if dist < 4.0 else 'Ionic'
                })

df_all = pd.DataFrame(all_interactions).sort_values('Distance').reset_index(drop=True)

print(f'Calculated {len(df_all)} pairwise distances')
print()
print('='*70)
```

## Section 5: Analyze Top Interactions

```python
print('='*70)
print('TOP 10 INTERACTIONS (Sidechain-Based)')
print('='*70)
print()

print(df_all.head(10).to_string(index=False))
print()
print('='*70)
print()

# Look for K93-D126
k93_d126 = df_all[
    ((df_all['NSP10_Res'] == 'LYS') & (df_all['NSP10_Num'] == 93) & 
     (df_all['NSP14_Res'] == 'ASP') & (df_all['NSP14_Num'] == 126))
]

print('K93-D126 VALIDATION:')
print()

if not k93_d126.empty:
    k93_row = k93_d126.iloc[0]
    rank = df_all.index[df_all['Distance'] == k93_row['Distance']].tolist()[0] + 1
    
    print(f'  Previous (Week 3): 2.78 Angstrom')
    print(f'  Current (Validated): {k93_row["Distance"]:.2f} Angstrom')
    print(f'  Atoms: {k93_row["NSP10_Atom"]} - {k93_row["NSP14_Atom"]}')
    print(f'  Rank: #{rank} out of {len(df_all)}')
    print(f'  Type: {k93_row["Type"]}')
    print()
    
    if rank == 1:
        print('  CONFIRMED: K93-D126 is #1 hot spot!')
        print('  Status: STRONGEST OVERALL across all structures!')
    else:
        top = df_all.iloc[0]
        print(f'  WARNING: K93-D126 is #{rank}, not #1')
        print(f'  #1 is: {top["NSP10_Res"]}{top["NSP10_Num"]}-{top["NSP14_Res"]}{top["NSP14_Num"]} ({top["Distance"]:.2f}A)')
else:
    print('  K93-D126 not found')
    print('  May be using different residue numbering')
    print('  Checking top interaction for K-D pairs...')

print()
print('='*70)
print()

# Check K76 involvement
print('K76 CONSERVATION CHECK:')
print()

k76_interactions = df_all[
    (df_all['NSP10_Res'] == 'LYS') & (df_all['NSP10_Num'] == 76)
]

if not k76_interactions.empty:
    print(f'  K76 has {len(k76_interactions)} interactions')
    print(f'  Strongest: {k76_interactions.iloc[0]["Distance"]:.2f} Angstrom')
    print(f'  Partner: {k76_interactions.iloc[0]["NSP14_Res"]}{k76_interactions.iloc[0]["NSP14_Num"]}')
    print()
    print('  Top 3 K76 interactions:')
    print(k76_interactions.head(3).to_string(index=False))
else:
    print('  K76 has NO interactions (or different numbering)')
    print('  Confirms: K76 NOT conserved in NSP10-NSP14')

print()
print('='*70)
```

## Section 6: Distance Matrix Heatmap

```python
print('Creating distance matrix heatmap...')
print()

df_all['NSP10_Label'] = df_all['NSP10_Res'] + df_all['NSP10_Num'].astype(str)
df_all['NSP14_Label'] = df_all['NSP14_Res'] + df_all['NSP14_Num'].astype(str)

df_close = df_all[df_all['Distance'] < 10.0].copy()

if len(df_close) > 0:
    matrix = df_close.pivot_table(
        index='NSP10_Label',
        columns='NSP14_Label',
        values='Distance',
        aggfunc='min'
    )
    
    plt.figure(figsize=(16, 8))
    sns.heatmap(
        matrix,
        cmap='RdYlGn_r',
        vmin=0,
        vmax=10,
        cbar_kws={'label': 'Distance (Angstrom)'},
        square=False
    )
    plt.title('7N0C: NSP10-NSP14 Distance Matrix (Sidechain)', fontsize=14, fontweight='bold')
    plt.xlabel('NSP14 Charged Residues', fontsize=12)
    plt.ylabel('NSP10 Charged Residues', fontsize=12)
    plt.tight_layout()
    plt.savefig(RESULTS_DIR / '7N0C_distance_matrix_comprehensive.png', dpi=300, bbox_inches='tight')
    plt.show()
    
    print('Heatmap saved')
else:
    print('No interactions < 10A to plot')
```

## Section 7: Visualize #1 Hot Spot

```python
if not df_all.empty:
    top = df_all.iloc[0]
    
    print('='*70)
    print('VISUALIZING #1 HOT SPOT')
    print('='*70)
    print()
    print(f'Interaction: {top["NSP10_Res"]}{top["NSP10_Num"]} <-> {top["NSP14_Res"]}{top["NSP14_Num"]}')
    print(f'Distance: {top["Distance"]:.2f} Angstrom')
    print(f'Atoms: {top["NSP10_Atom"]} - {top["NSP14_Atom"]}')
    print()
    
    with open(PDB_FILE, 'r') as f:
        pdb_data = f.read()
    
    view = py3Dmol.view(width=800, height=600)
    view.addModel(pdb_data, 'pdb')
    
    view.setStyle({'chain': nsp10_chain}, {'cartoon': {'color': 'cyan', 'opacity': 0.3}})
    view.setStyle({'chain': nsp14_chain}, {'cartoon': {'color': 'orange', 'opacity': 0.3}})
    
    view.addStyle({'chain': nsp10_chain, 'resi': int(top['NSP10_Num'])},
                  {'sphere': {'color': 'blue', 'radius': 1.0}})
    view.addStyle({'chain': nsp14_chain, 'resi': int(top['NSP14_Num'])},
                  {'sphere': {'color': 'red', 'radius': 1.0}})
    
    view.addLabel(f"{top['NSP10_Res']}{int(top['NSP10_Num'])} (NSP10)",
                  {'fontColor': 'white', 'fontSize': 16, 'backgroundColor': 'blue', 'backgroundOpacity': 0.8},
                  {'chain': nsp10_chain, 'resi': int(top['NSP10_Num'])})
    
    view.addLabel(f"{top['NSP14_Res']}{int(top['NSP14_Num'])} {top['Distance']:.2f}A",
                  {'fontColor': 'white', 'fontSize': 16, 'backgroundColor': 'red', 'backgroundOpacity': 0.8},
                  {'chain': nsp14_chain, 'resi': int(top['NSP14_Num'])})
    
    view.zoomTo({'chain': nsp10_chain, 'resi': int(top['NSP10_Num'])})
    view.setBackgroundColor('white')
    
    view.show()
```

## Section 8: Export Results

```python
print('='*70)
print('EXPORTING RESULTS')
print('='*70)
print()

csv_file = RESULTS_DIR / '7N0C_comprehensive_validation_all.csv'
df_all.to_csv(csv_file, index=False)
print(f'All interactions: {csv_file}')

top20_file = RESULTS_DIR / '7N0C_comprehensive_validation_top20.csv'
df_all.head(20).to_csv(top20_file, index=False)
print(f'Top 20: {top20_file}')

summary_data = {
    'pdb_id': '7N0C',
    'analysis_type': 'comprehensive_validation',
    'method': 'sidechain_based',
    'total_interactions': len(df_all),
    'strongest': {
        'residues': f"{df_all.iloc[0]['NSP10_Res']}{df_all.iloc[0]['NSP10_Num']}-{df_all.iloc[0]['NSP14_Res']}{df_all.iloc[0]['NSP14_Num']}",
        'distance': float(df_all.iloc[0]['Distance']),
        'atoms': f"{df_all.iloc[0]['NSP10_Atom']}-{df_all.iloc[0]['NSP14_Atom']}"
    }
}

json_file = RESULTS_DIR / '7N0C_comprehensive_validation.json'
with open(json_file, 'w') as f:
    json.dump(summary_data, f, indent=2)
print(f'Summary: {json_file}')

print()
print('All results exported!')
print('='*70)
```
