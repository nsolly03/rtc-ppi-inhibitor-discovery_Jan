## Setup

```python
import warnings
warnings.filterwarnings('ignore')

import nglview as nv
from Bio import PDB
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Set style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

# Paths
PDB_DIR = Path('../data/structures/pdb')
RESULTS_DIR = Path('../data/analysis_results')
RESULTS_DIR.mkdir(exist_ok=True, parents=True)

print("✓ Setup complete")
print(f"  PDB directory: {PDB_DIR}")
print(f"  Results directory: {RESULTS_DIR}")
```

## Structure 1: 6W4H (NSP10-NSP16) ✅ PRIMARY ANALYSIS COMPLETE

```python
print("="*70)
print("STRUCTURE 1: 6W4H (NSP10-NSP16)")
print("="*70)
print()
print("Function: 2'-O-methyltransferase complex")
print("Role: Modifies viral RNA to evade immune detection")
print("Status: ✅ COMPLETE ANALYSIS (Week 1-2)")
print()

# Load structure
parser = PDB.PDBParser(QUIET=True)
structure_6W4H = parser.get_structure('6W4H', str(PDB_DIR / '6W4H.pdb'))

# Chain information
print("CHAINS:")
for chain in structure_6W4H[0]:
    residues = [r for r in chain.get_residues() if r.id[0] == ' ']
    print(f"  Chain {chain.id}: {len(residues)} residues")
print()

# Known analysis results
print("INTERFACE ANALYSIS (Already complete):")
print("  Interface: NSP10 (Chain B) - NSP16 (Chain A)")
print("  Hot spot: K76-D107 (5.15 Å salt bridge)")
print("  Charged triad: K76-K78-D107")
print("  Interface residues: 22 total (16 NSP10, 6 NSP16)")
print("  Grid box center: (75.883, 11.641, 10.087)")
print("  Grid box size: 25 × 25 × 25 Ų")
print()
print("NEXT STEP: fpocket to identify binding pockets")
print()

# Visualize
view = nv.show_file(str(PDB_DIR / '6W4H.pdb'))
view.clear_representations()
view.add_cartoon('protein', color='chainid')
view.add_label('Chain B', labelType='text', labelText='NSP10', 
               color='blue', backgroundColor='white', backgroundOpacity=0.8)
view.add_label('Chain A', labelType='text', labelText='NSP16',
               color='red', backgroundColor='white', backgroundOpacity=0.8)
view.center()
view.camera = 'orthographic'
view
```

## Structure 2: 7DFG (NSP12-NSP7-NSP8) ⭐ HIGHEST PRIORITY

```python
print("="*70)
print("STRUCTURE 2: 7DFG (NSP12-NSP7-NSP8)")
print("="*70)
print()
print("Function: RNA-dependent RNA polymerase holoenzyme")
print("Role: Core viral replication machinery")
print("Priority: ⭐ HIGHEST - essential for viral replication")
print()

# Load structure
structure_7DFG = parser.get_structure('7DFG', str(PDB_DIR / '7DFG.pdb'))

# Chain information
print("CHAINS:")
chain_info_7DFG = {}
for chain in structure_7DFG[0]:
    residues = [r for r in chain.get_residues() if r.id[0] == ' ']
    chain_info_7DFG[chain.id] = len(residues)
    print(f"  Chain {chain.id}: {len(residues)} residues")
print()

# Identify which chain is which protein
print("PROTEIN IDENTIFICATION:")
for chain_id, num_residues in chain_info_7DFG.items():
    if num_residues > 700:  # NSP12 is large (~930 residues)
        print(f"  Chain {chain_id}: NSP12 (RdRp) - {num_residues} residues")
    elif 70 < num_residues < 90:  # NSP7 ~83 residues
        print(f"  Chain {chain_id}: NSP7 (cofactor) - {num_residues} residues")
    elif 180 < num_residues < 210:  # NSP8 ~198 residues
        print(f"  Chain {chain_id}: NSP8 (cofactor) - {num_residues} residues")
    else:
        print(f"  Chain {chain_id}: Unknown - {num_residues} residues")
print()

print("POTENTIAL INTERFACES TO TARGET:")
print("  1. NSP12-NSP7 (polymerase-cofactor)")
print("  2. NSP12-NSP8 (polymerase-cofactor)")
print("  3. NSP7-NSP8 (cofactor-cofactor) - optional")
print()
print("ANALYSIS NEEDED:")
print("  [ ] Map interface residues for each PPI")
print("  [ ] Identify hot spots")
print("  [ ] Run fpocket")
print("  [ ] Select best interface(s) to target")
print()

# Visualize
view = nv.show_file(str(PDB_DIR / '7DFG.pdb'))
view.clear_representations()
view.add_cartoon('protein', color='chainid')
view.center()
view.camera = 'orthographic'
view
```

## Summary Comparison Table

```python
# Create comprehensive comparison
summary_data = []

structures = {
    '6W4H': ('NSP10-NSP16', '2\'-O-MTase', 'High', 'Immune evasion', '✅ Complete'),
    '7DFG': ('NSP12-NSP7-NSP8', 'RdRp holoenzyme', '⭐ Highest', 'Replication core', '⏳ Pending'),
    '6XEZ': ('NSP12-NSP7-NSP8', 'RdRp (alternative)', 'High', 'Comparison', '⏳ Pending'),
    '7EDI': ('NSP10-NSP14', 'ExoN complex', 'High', 'Proofreading', '⏳ Pending'),
    '6W9C': ('NSP13', 'Helicase', 'Medium', 'RNA unwinding', '⏳ Pending')
}

for pdb_id, (complex_name, function, priority, role, status) in structures.items():
    # Load and count chains
    structure = parser.get_structure(pdb_id, str(PDB_DIR / f'{pdb_id}.pdb'))
    chains = list(structure[0].get_chains())
    num_chains = len(chains)
    
    # Count total residues
    total_residues = sum([len([r for r in chain.get_residues() if r.id[0] == ' ']) 
                         for chain in chains])
    
    # Determine number of PPIs
    if 'NSP12-NSP7-NSP8' in complex_name:
        num_ppis = '2-3'  # NSP12-NSP7, NSP12-NSP8, optional NSP7-NSP8
    elif num_chains >= 2:
        num_ppis = '1'
    else:
        num_ppis = '0 (single protein)'
    
    summary_data.append({
        'PDB': pdb_id,
        'Complex': complex_name,
        'Function': function,
        'Chains': num_chains,
        'Total_Residues': total_residues,
        'PPIs': num_ppis,
        'Priority': priority,
        'Role': role,
        'Analysis_Status': status
    })

summary_df = pd.DataFrame(summary_data)

print("="*100)
print("COMPREHENSIVE STRUCTURE SUMMARY")
print("="*100)
print()
print(summary_df.to_string(index=False))
print()

# Save
summary_df.to_csv(RESULTS_DIR / 'structure_overview_summary.csv', index=False)
print(f"✓ Summary saved to: {RESULTS_DIR / 'structure_overview_summary.csv'}")
```
