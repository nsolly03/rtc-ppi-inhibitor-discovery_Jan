## Section 1: Setup and Configuration

```python
import warnings
warnings.filterwarnings('ignore')

import py3Dmol
from Bio import PDB
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Plotting style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette('husl')

# Paths
PDB_DIR = Path('data/structures/pdb')
RESULTS_DIR = Path('data/analysis_results')
PDB_FILE = PDB_DIR / '7DFG.pdb'

# Parse structure
parser = PDB.PDBParser(QUIET=True)
structure = parser.get_structure('7DFG', str(PDB_FILE))
model = structure[0]

print("="*70)
print("7DFG COMPREHENSIVE HOT SPOT VALIDATION")
print("="*70)
print()
print("✓ Structure loaded")
print(f"  PDB: {PDB_FILE}")
print(f"  Chains: {[c.id for c in model]}")
print()
print("Target interfaces:")
print("  1. NSP12-NSP7 (Chain A - Chain C)")
print("  2. NSP12-NSP8 (Chain A - Chains B,G)")
print()
print("Method: ALL charged residue pairwise distances")
print("="*70)
```

## Section 2A: NSP12-NSP7 - Find ALL Charged Residues

```python
# Chain assignments (from discovery notebook)
nsp12_chain = 'A'
nsp7_chain = 'C'

print("="*70)
print("NSP12-NSP7 INTERFACE: Charged Residue Inventory")
print("="*70)
print()

# Find ALL positively charged in NSP12
nsp12_positive = []
for res in model[nsp12_chain]:
    if res.id[0] == ' ' and res.get_resname() in ['LYS', 'ARG']:
        if 'CA' in res:
            nsp12_positive.append({
                'residue': res,
                'type': res.get_resname(),
                'pdb_num': res.id[1],
                'ca_coord': res['CA'].get_coord(),
                'label': f"{res.get_resname()[0]}{res.id[1]}"
            })

# Find ALL negatively charged in NSP7
nsp7_negative = []
for res in model[nsp7_chain]:
    if res.id[0] == ' ' and res.get_resname() in ['ASP', 'GLU']:
        if 'CA' in res:
            nsp7_negative.append({
                'residue': res,
                'type': res.get_resname(),
                'pdb_num': res.id[1],
                'ca_coord': res['CA'].get_coord(),
                'label': f"{res.get_resname()[0]}{res.id[1]}"
            })

print(f"NSP12 (Chain {nsp12_chain}) Positive Charges:")
print(f"  Lysines (LYS): {len([r for r in nsp12_positive if r['type'] == 'LYS'])}")
print(f"  Arginines (ARG): {len([r for r in nsp12_positive if r['type'] == 'ARG'])}")
print(f"  TOTAL: {len(nsp12_positive)}")
print()
print(f"NSP7 (Chain {nsp7_chain}) Negative Charges:")
print(f"  Aspartates (ASP): {len([r for r in nsp7_negative if r['type'] == 'ASP'])}")
print(f"  Glutamates (GLU): {len([r for r in nsp7_negative if r['type'] == 'GLU'])}")
print(f"  TOTAL: {len(nsp7_negative)}")
print()
print(f"📊 TOTAL COMBINATIONS: {len(nsp12_positive)} × {len(nsp7_negative)} = {len(nsp12_positive) * len(nsp7_negative)}")
print()
print("="*70)
```
